```python
import json
import math

import requests
import numpy as np
import matplotlib.pyplot as plt

from scipy.integrate import solve_ivp
from scipy.interpolate import PchipInterpolator
from scipy.signal import savgol_filter


# ============================================================
# Physical parameters
# ============================================================

rider_mass = 64.0
trike_mass = 20.0
backpack_mass = 5.0

m = rider_mass + trike_mass + backpack_mass

g = 9.81
rho = 1.225

# Rolling-resistance coefficients to test.
Crr_values = np.linspace(0.003, 0.012, 4)

# Aerodynamic drag areas to test, in m^2.
CdA_values = np.linspace(0.20, 0.80, 4)

initial_speed_kmh = 0.0
initial_speed = initial_speed_kmh / 3.6


# ============================================================
# Geocoding
# ============================================================

def geocode(address):
    url = "https://nominatim.openstreetmap.org/search"

    params = {
        "q": address,
        "format": "json",
        "countrycodes": "au",
        "limit": 10,
    }

    headers = {
        "User-Agent": "Castle-Hill-route-analysis/1.0"
    }

    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=30,
    )
    response.raise_for_status()

    results = response.json()

    if not results:
        raise RuntimeError(
            f"No geocoding results for {address!r}"
        )

    print(f"\nNominatim candidates for {address!r}:")

    for i, result in enumerate(results):
        print(
            f"{i}: "
            f"{result.get('display_name')} "
            f"({result.get('lat')}, {result.get('lon')})"
        )

    for result in results:
        display_name = result.get("display_name", "").lower()

        if "stanley street" in display_name:
            return (
                float(result["lon"]),
                float(result["lat"]),
            )

    raise RuntimeError(
        f"Could not find a geocoding result containing "
        f"Stanley Street for {address!r}"
    )


start_lon, start_lat = geocode(
    "320 Stanley Street, North Ward, Queensland, Australia"
)

# Summit of Castle Hill.
finish_lon = 146.8026
finish_lat = -19.2580


# ============================================================
# OSRM route
# ============================================================

osrm_url = (
    "https://router.project-osrm.org/route/v1/driving/"
    f"{start_lon},{start_lat};{finish_lon},{finish_lat}"
)

osrm_params = {
    "overview": "full",
    "geometries": "geojson",
    "steps": "true",
}

response = requests.get(
    osrm_url,
    params=osrm_params,
    timeout=60,
)
response.raise_for_status()

route_data = response.json()

if route_data.get("code") != "Ok":
    raise RuntimeError(
        f"OSRM routing failed: {route_data}"
    )

route = route_data["routes"][0]

route_coords = route["geometry"]["coordinates"]
route_steps = route.get("legs", [])[0].get("steps", [])


# ============================================================
# Check that OSRM selected the expected Castle Hill route
# ============================================================

step_names = [
    step.get("name", "").strip()
    for step in route_steps
]

print("\nOSRM route steps:")

for name in step_names:
    print(f"  {name}")

required_streets = [
    "Gregory Street",
    "Stanley Street",
    "Castle Hill Road",
]

missing_streets = [
    street
    for street in required_streets
    if not any(
        street.lower() in name.lower()
        for name in step_names
    )
]

if missing_streets:
    raise RuntimeError(
        "OSRM route does not contain all expected streets. "
        f"Missing: {missing_streets}"
    )


# ============================================================
# Haversine distance
# ============================================================

def haversine(lon1, lat1, lon2, lat2):
    R = 6_371_000.0

    phi1 = math.radians(lat1)
    phi2 = math.radians(lat2)

    dphi = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)

    a = (
        math.sin(dphi / 2.0) ** 2
        + math.cos(phi1)
        * math.cos(phi2)
        * math.sin(dlambda / 2.0) ** 2
    )

    return 2.0 * R * math.asin(math.sqrt(a))


route_distance = np.zeros(
    len(route_coords),
    dtype=float,
)

for i in range(1, len(route_coords)):
    lon1, lat1 = route_coords[i - 1]
    lon2, lat2 = route_coords[i]

    route_distance[i] = (
        route_distance[i - 1]
        + haversine(lon1, lat1, lon2, lat2)
    )

print(
    f"\nOSRM route length: "
    f"{route['distance'] / 1000:.3f} km"
)

print(
    f"OSRM route duration: "
    f"{route['duration'] / 60:.1f} min"
)

if not 2000.0 <= route_distance[-1] <= 4000.0:
    raise RuntimeError(
        f"Unexpected Castle Hill route length: "
        f"{route_distance[-1]:.1f} m"
    )


# ============================================================
# Resample route at approximately 10 m spacing
# ============================================================

spacing = 10.0

s_regular = np.arange(
    0.0,
    route_distance[-1],
    spacing,
)

if (
    len(s_regular) == 0
    or s_regular[-1] < route_distance[-1]
):
    s_regular = np.append(
        s_regular,
        route_distance[-1],
    )

route_lons = np.array(
    [point[0] for point in route_coords],
    dtype=float,
)

route_lats = np.array(
    [point[1] for point in route_coords],
    dtype=float,
)

lons = np.interp(
    s_regular,
    route_distance,
    route_lons,
)

lats = np.interp(
    s_regular,
    route_distance,
    route_lats,
)

print(
    f"Resampled route: {len(s_regular)} points"
)

print(
    f"Resampled route distance: "
    f"{s_regular[-1] / 1000:.3f} km"
)


# ============================================================
# Queensland Government QldDem
# ============================================================

dem_url = (
    "https://spatial-img.information.qld.gov.au/"
    "arcgis/rest/services/Elevation/QldDem/"
    "ImageServer/getSamples"
)



def query_dem_points(lons, lats, chunk_size=10):
    import json
    import time

    import numpy as np
    import requests

    dem_url = (
        "https://spatial-img.information.qld.gov.au/"
        "arcgis/rest/services/Elevation/QldDem/"
        "ImageServer/getSamples"
    )

    lons = np.asarray(lons, dtype=float)
    lats = np.asarray(lats, dtype=float)

    if lons.ndim != 1 or lats.ndim != 1:
        raise ValueError("lons and lats must be one-dimensional")

    if len(lons) != len(lats):
        raise ValueError("lons and lats must have equal lengths")

    if not np.all(np.isfinite(lons)) or not np.all(np.isfinite(lats)):
        raise ValueError("Coordinates must all be finite")

    if chunk_size < 1:
        raise ValueError("chunk_size must be positive")

    elevations = np.full(len(lons), np.nan, dtype=float)

    headers = {
        "User-Agent": "Castle-Hill-route-analysis/1.0"
    }

    session = requests.Session()
    session.headers.update(headers)

    def request_samples(indices):
        points = [
            [float(lons[i]), float(lats[i])]
            for i in indices
        ]

        geometry = {
            "points": points,
            "spatialReference": {"wkid": 4326},
        }

        params = {
            "geometry": json.dumps(geometry),
            "geometryType": "esriGeometryMultipoint",
            "returnFirstValueOnly": "true",
            "interpolation": "RSP_BilinearInterpolation",
            "outFields": "*",
            "f": "json",
        }

        response = session.get(
            dem_url,
            params=params,
            timeout=60,
        )
        response.raise_for_status()

        data = response.json()

        if "error" in data:
            raise RuntimeError(
                f"QldDem API error: {data['error']}"
            )

        samples = data.get("samples")

        if samples is None:
            raise RuntimeError(
                f"Response has no samples: {data}"
            )

        if len(samples) != len(indices):
            raise RuntimeError(
                f"Expected {len(indices)} samples; "
                f"received {len(samples)}"
            )

        values = []

        for sample in samples:
            value = sample.get("value")

            try:
                value = float(value)
            except (TypeError, ValueError):
                value = np.nan

            values.append(value)

        return np.asarray(values, dtype=float)

    try:
        for start in range(0, len(lons), chunk_size):
            end = min(start + chunk_size, len(lons))
            indices = list(range(start, end))

            try:
                elevations[start:end] = request_samples(indices)

            except Exception as batch_error:
                print(
                    f"Batch {start}:{end} failed: "
                    f"{batch_error}"
                )
                print("Retrying this batch point by point...")

                for i in indices:
                    last_error = None

                    for attempt in range(3):
                        try:
                            elevations[i] = request_samples([i])[0]
                            last_error = None
                            break

                        except Exception as exc:
                            last_error = exc

                            if attempt < 2:
                                time.sleep(2 ** attempt)

                    if last_error is not None:
                        raise RuntimeError(
                            f"Could not retrieve elevation at "
                            f"point {i}: "
                            f"longitude={lons[i]:.8f}, "
                            f"latitude={lats[i]:.8f}. "
                            f"Last error: {last_error}"
                        ) from last_error

                    time.sleep(0.1)

            print(
                f"Retrieved elevations for "
                f"{end}/{len(lons)} points"
            )

    finally:
        session.close()

    return elevations



print("\nQuerying QldDem...")

h_ascent = query_dem_points(
    lons,
    lats,
)

if np.any(~np.isfinite(h_ascent)):
    bad = np.count_nonzero(
        ~np.isfinite(h_ascent)
    )

    raise RuntimeError(
        f"QldDem returned {bad} invalid elevation values"
    )


# ============================================================
# Basic DEM sanity checks
# ============================================================

if np.nanmax(h_ascent) <= 150.0:
    raise RuntimeError(
        "DEM maximum elevation is unexpectedly low: "
        f"{np.nanmax(h_ascent):.1f} m"
    )

if (
    np.nanmax(h_ascent)
    - np.nanmin(h_ascent)
    <= 100.0
):
    raise RuntimeError(
        "DEM elevation range is unexpectedly small: "
        f"{np.nanmax(h_ascent) - np.nanmin(h_ascent):.1f} m"
    )


# ============================================================
# Smooth elevation
# ============================================================

window_length = 11

if len(h_ascent) < window_length:
    raise RuntimeError(
        "Not enough elevation points for "
        f"a Savitzky-Golay window of {window_length}"
    )

if window_length % 2 == 0:
    raise RuntimeError(
        "Savitzky-Golay window length must be odd"
    )

h_ascent = savgol_filter(
    h_ascent,
    window_length=window_length,
    polyorder=3,
)


# ============================================================
# Reverse route: summit -> Gregory Street
# ============================================================

total_distance = s_regular[-1]

s = total_distance - s_regular[::-1]
s = s - s[0]

h = h_ascent[::-1]


# ============================================================
# Elevation summary
# ============================================================

summit_elevation = float(h[0])
gregory_elevation = float(h[-1])

elevation_drop = (
    summit_elevation
    - gregory_elevation
)

mean_gradient = (
    elevation_drop / s[-1]
)

print(
    "\n===== Castle Hill elevation summary ====="
)

print(
    f"Route distance: "
    f"{s[-1] / 1000:.3f} km"
)

print(
    f"Summit elevation: "
    f"{summit_elevation:.1f} m"
)

print(
    f"Gregory Street elevation: "
    f"{gregory_elevation:.1f} m"
)

print(
    f"Elevation drop: "
    f"{elevation_drop:.1f} m"
)

print(
    f"Mean downhill gradient: "
    f"{100.0 * mean_gradient:.2f}%"
)

if not 150.0 <= summit_elevation <= 350.0:
    raise RuntimeError(
        f"Unexpected summit elevation: "
        f"{summit_elevation:.1f} m"
    )

if not 100.0 <= elevation_drop <= 350.0:
    raise RuntimeError(
        f"Unexpected elevation drop: "
        f"{elevation_drop:.1f} m"
    )


# ============================================================
# Elevation graph
# ============================================================

plt.figure(figsize=(11, 5))

plt.plot(
    s / 1000.0,
    h,
)

plt.xlabel("Distance from summit (km)")
plt.ylabel("Elevation (m)")
plt.title(
    "Castle Hill Road \u2014 elevation profile"
)

plt.grid(True)
plt.tight_layout()
plt.show()


# ============================================================
# Gradient
# ============================================================

dhds = np.gradient(
    h,
    s,
)

# s increases downhill, so dh/ds = -sin(theta).
sin_theta = np.clip(
    -dhds,
    -1.0,
    1.0,
)

cos_theta = np.sqrt(
    np.maximum(
        0.0,
        1.0 - sin_theta ** 2,
    )
)


# ============================================================
# Gradient graph
# ============================================================

plt.figure(figsize=(11, 5))

plt.plot(
    s / 1000.0,
    100.0 * sin_theta,
)

plt.axhline(
    0.0,
    linestyle="--",
)

plt.xlabel("Distance from summit (km)")
plt.ylabel("Downhill gradient (%)")
plt.title(
    "Castle Hill Road \u2014 gradient profile"
)

plt.grid(True)
plt.tight_layout()
plt.show()


# ============================================================
# Interpolators for the ODE
# ============================================================

sin_theta_fn = PchipInterpolator(
    s,
    sin_theta,
    extrapolate=True,
)

cos_theta_fn = PchipInterpolator(
    s,
    cos_theta,
    extrapolate=True,
)


# ============================================================
# Speed calculation
#
# Solve for w = v^2:
#
# dw/ds = 2 * (
#     g sin(theta)
#     - Crr g cos(theta)
#     - rho CdA/(2m) * w
# )
# ============================================================

def calculate_speed_profile(CdA, Crr):
    drag_coefficient = rho * CdA / (2.0 * m)

    def ode(distance, w):
        st = float(sin_theta_fn(distance))
        ct = float(cos_theta_fn(distance))

        kinetic_term = max(
            float(w[0]),
            0.0,
        )

        acceleration = (
            g * st
            - Crr * g * ct
            - drag_coefficient * kinetic_term
        )

        # Prevent squared speed becoming negative.
        if (
            kinetic_term <= 0.0
            and acceleration < 0.0
        ):
            return [0.0]

        return [2.0 * acceleration]

    solution = solve_ivp(
        ode,
        (
            float(s[0]),
            float(s[-1]),
        ),
        [initial_speed ** 2],
        t_eval=s,
        rtol=1e-9,
        atol=1e-11,
        max_step=2.0,
    )

    if not solution.success:
        raise RuntimeError(
            f"Speed integration failed for "
            f"CdA={CdA}, Crr={Crr}: "
            f"{solution.message}"
        )

    w = np.maximum(
        solution.y[0],
        0.0,
    )

    return np.sqrt(w)


# ============================================================
# Calculate speed profiles for every parameter combination
# ============================================================

speed_profiles = {}

total_combinations = (
    len(Crr_values) * len(CdA_values)
)

print(
    f"\nCalculating {total_combinations} parameter combinations..."
)

for Crr in Crr_values:
    for CdA in CdA_values:
        key = (
            float(Crr),
            float(CdA),
        )

        speed_profiles[key] = calculate_speed_profile(
            float(CdA),
            float(Crr),
        )

        print(
            f"  {len(speed_profiles):>3}/{total_combinations}: "
            f"Crr={Crr:.4f}, CdA={CdA:.2f}"
        )


# ============================================================
# Speed graph: all combinations
# ============================================================

plt.figure(figsize=(13, 7))

for (Crr, CdA), v in speed_profiles.items():
    plt.plot(
        s / 1000.0,
        v * 3.6,
        label=f"Crr={Crr:.3f}, CdA={CdA:.2f}",
    )

plt.xlabel("Distance from summit (km)")
plt.ylabel("Speed (km/h)")
plt.title(
    "Castle Hill Road \u2014 theoretical brake-failure speed"
)

plt.grid(True)
plt.legend(
    ncol=2,
    fontsize="small",
)
plt.tight_layout()
plt.savefig("Castle hill descent speed constant gravity.svg")
plt.show()


# ============================================================
# Trip duration
#
# Only intervals whose two endpoints both have v > 0
# contribute to the calculated coasting time.
# ============================================================

def calculate_time(s, v):
    s = np.asarray(
        s,
        dtype=float,
    )

    v = np.asarray(
        v,
        dtype=float,
    )

    if len(s) != len(v):
        raise ValueError(
            "s and v must have the same length"
        )

    if len(s) < 2:
        return 0.0

    ds = np.diff(s)

    valid_intervals = (
        np.isfinite(v[:-1])
        & np.isfinite(v[1:])
        & (v[:-1] > 0.0)
        & (v[1:] > 0.0)
    )

    dt = np.zeros(
        len(ds),
        dtype=float,
    )

    dt[valid_intervals] = (
        0.5
        * (
            1.0 / v[:-1][valid_intervals]
            + 1.0 / v[1:][valid_intervals]
        )
        * ds[valid_intervals]
    )

    return float(np.sum(dt))


# ============================================================
# Results for every combination
# ============================================================

print(
    "\n===== Brake-failure speed results ====="
)

summary_rows = []

for (Crr, CdA), v in speed_profiles.items():
    duration = calculate_time(
        s,
        v,
    )

    positive_points = (
        np.isfinite(v)
        & (v > 0.0)
    )

    positive_intervals = (
        np.isfinite(v[:-1])
        & np.isfinite(v[1:])
        & (v[:-1] > 0.0)
        & (v[1:] > 0.0)
    )

    coasting_distance = float(
        np.sum(
            np.diff(s)[positive_intervals]
        )
    )

    zero_or_invalid_distance = (
        float(s[-1] - s[0])
        - coasting_distance
    )

    print(
        f"\nCrr={Crr:.4f}, CdA={CdA:.2f} m^2"
    )

    if np.any(positive_points):
        max_speed = float(
            np.max(v[positive_points]) * 3.6
        )

        final_speed = float(v[-1] * 3.6)

        print(
            f"  final speed: {final_speed:.1f} km/h"
        )

        print(
            f"  maximum speed: {max_speed:.1f} km/h"
        )

        print(
            f"  coasting distance: "
            f"{coasting_distance / 1000:.3f} km"
        )

        print(
            f"  zero/invalid-speed distance: "
            f"{zero_or_invalid_distance / 1000:.3f} km"
        )

        print(
            f"  coasting time: {duration / 60:.2f} min"
        )

    else:
        max_speed = 0.0
        final_speed = 0.0

        print(
            "  speed is zero/non-finite throughout"
        )

        print(
            "  coasting time: 0.00 min"
        )

    summary_rows.append({
        "Crr": float(Crr),
        "CdA (m^2)": float(CdA),
        "Maximum speed (km/h)": max_speed,
        "Final speed (km/h)": final_speed,
        "Coasting distance (km)": (
            coasting_distance / 1000.0
        ),
        "Coasting time (min)": duration / 60.0,
        "Zero/invalid distance (km)": (
            zero_or_invalid_distance / 1000.0
        ),
    })


# ============================================================
# Compact summary table
# ============================================================

summary_rows.sort(
    key=lambda row: (
        row["Crr"],
        row["CdA (m^2)"],
    )
)

print("\n===== Compact results table =====")

header = (
    f"{'Crr':>8} "
    f"{'CdA (m^2)':>11} "
    f"{'Max speed':>13} "
    f"{'Final speed':>13} "
    f"{'Distance (km)':>15} "
    f"{'Time (min)':>12}"
)

print(header)
print("-" * len(header))

for row in summary_rows:
    print(
        f"{row['Crr']:8.4f} "
        f"{row['CdA (m^2)']:11.2f} "
        f"{row['Maximum speed (km/h)']:13.1f} "
        f"{row['Final speed (km/h)']:13.1f} "
        f"{row['Coasting distance (km)']:15.3f} "
        f"{row['Coasting time (min)']:12.2f}"
    )
```

```python

# Castle Hill Road descent: close-up 3D terrain animation with elevation bands.
#
# Features:
#   - No visible axes, ticks, grid lines or axis labels.
#   - Discrete, fixed elevation bands and a matching legend.
#   - Moving camera that follows the tricycle.
#   - Close-up view with exaggerated vertical relief.
#   - Three-wheel tricycle schematic with frame and rider.
#   - Route and recent-travel trail.
#   - Simulation time, velocity, distance and current elevation.
#   - Frame number and percentage progress during rendering.
#   - Parallel rendering using up to eight CPU cores.
#   - MP4 output.
#
# Required existing notebook variables:
#   terrain_x, terrain_y, terrain_z
#   road_x, road_y, road_z
#   distance, velocity
#
# Optional:
#   h -- elevation profile ordered summit -> Gregory Street.

import os
import shutil
import subprocess
import tempfile
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor, as_completed

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.colors import BoundaryNorm, ListedColormap
from matplotlib.patches import Patch

# ---------------------------------------------------------------------
# Configuration
# ---------------------------------------------------------------------

OUTPUT_FILE = "castle_hill_terrain_descent.mp4"

MAX_WORKERS = min(8, os.cpu_count() or 1)
VIDEO_FPS = 10
VIDEO_DPI = 100
VIDEO_CRF = 18
TERRAIN_STRIDE = 2

CAMERA_WIDTH = 700.0
CAMERA_DEPTH = 550.0
CAMERA_LOOK_AHEAD = 90.0
CAMERA_ELEVATION = 38.0
CAMERA_TURN_SMOOTHING = 0.20

LOCAL_Z_BELOW = 65.0
LOCAL_Z_ABOVE = 95.0
VERTICAL_ASPECT_FACTOR = 2.5

TRIKE_SCALE = 4.0
WHEEL_RADIUS = 0.34 * TRIKE_SCALE
REAR_TRACK = 0.85 * TRIKE_SCALE
REAR_OFFSET = -0.85 * TRIKE_SCALE
FRONT_OFFSET = 0.90 * TRIKE_SCALE

# Fixed elevation intervals in metres above the DEM vertical datum.
# The colour scale and legend do not change as the camera moves.
ELEVATION_BANDS = [
    (0, 50, "#2166ac", "0–50 m"),
    (50, 100, "#67a9cf", "50–100 m"),
    (100, 150, "#d1e5f0", "100–150 m"),
    (150, 200, "#fddbc7", "150–200 m"),
    (200, 250, "#ef8a62", "200–250 m"),
    (250, 300, "#b2182b", "250–300 m"),
]

# ---------------------------------------------------------------------
# Validate and prepare data
# ---------------------------------------------------------------------

required = (
    "terrain_x", "terrain_y", "terrain_z",
    "road_x", "road_y", "road_z",
    "distance", "velocity",
)
missing = [name for name in required if name not in globals()]
if missing:
    raise NameError("Missing notebook variables: " + ", ".join(missing))

tx = np.asarray(terrain_x, dtype=float)
ty = np.asarray(terrain_y, dtype=float)
tz = np.asarray(terrain_z, dtype=float)

rx = np.asarray(road_x, dtype=float).ravel()
ry = np.asarray(road_y, dtype=float).ravel()
rz = np.asarray(road_z, dtype=float).ravel()

dist = np.asarray(distance, dtype=float).ravel()
vel = np.asarray(velocity, dtype=float).ravel()

if tx.ndim != 2 or tx.shape != ty.shape or tx.shape != tz.shape:
    raise ValueError("terrain_x, terrain_y and terrain_z must have matching 2D shapes.")
if not (len(rx) == len(ry) == len(rz)):
    raise ValueError("road_x, road_y and road_z must have equal lengths.")
if len(rx) < 2 or len(dist) < 2 or len(dist) != len(vel):
    raise ValueError("The route and simulation arrays are incomplete.")

valid = np.isfinite(dist) & np.isfinite(vel)
dist, vel = dist[valid], vel[valid]
order = np.argsort(dist)
dist, vel = dist[order], vel[order]
dist, unique_idx = np.unique(dist, return_index=True)
vel = vel[unique_idx]

if len(dist) < 2 or dist[-1] <= dist[0]:
    raise ValueError("Simulation distance must be strictly increasing.")
if not (
    np.all(np.isfinite(rx))
    and np.all(np.isfinite(ry))
    and np.all(np.isfinite(rz))
):
    raise ValueError("The road arrays contain non-finite coordinates.")

# Orient the route summit -> Gregory Street using h when available.
reverse_route = False
if "h" in globals():
    href = np.asarray(h, dtype=float).ravel()
    href = href[np.isfinite(href)]
    if len(href) >= 2:
        same_error = abs(rz[0] - href[0]) + abs(rz[-1] - href[-1])
        reverse_error = abs(rz[-1] - href[0]) + abs(rz[0] - href[-1])
        reverse_route = reverse_error < same_error
    else:
        reverse_route = rz[0] < rz[-1]
else:
    reverse_route = rz[0] < rz[-1]

if reverse_route:
    rx, ry, rz = rx[::-1].copy(), ry[::-1].copy(), rz[::-1].copy()
    print("Reversed road coordinates to match the descent direction.")

route_ds = np.hypot(np.diff(rx), np.diff(ry))
route_s = np.concatenate(([0.0], np.cumsum(route_ds)))
route_s, route_unique_idx = np.unique(route_s, return_index=True)
rx, ry, rz = rx[route_unique_idx], ry[route_unique_idx], rz[route_unique_idx]

route_length = float(route_s[-1])
simulation_length = float(dist[-1] - dist[0])
if route_length <= 0 or simulation_length <= 0:
    raise ValueError("Could not determine a positive route or simulation length.")

route_progress = (dist - dist[0]) / simulation_length * route_length
path_x = np.interp(route_progress, route_s, rx)
path_y = np.interp(route_progress, route_s, ry)
path_z = np.interp(route_progress, route_s, rz)

# Integrate dt = ds/v to calculate elapsed simulation time.
safe_velocity = np.maximum(vel, 0.1)
elapsed = np.zeros_like(dist)
elapsed[1:] = np.cumsum(
    np.diff(dist) * 0.5 * (
        1.0 / safe_velocity[:-1] + 1.0 / safe_velocity[1:]
    )
)

duration = float(elapsed[-1])
if not np.isfinite(duration) or duration <= 0:
    raise ValueError("Could not calculate a valid simulation duration.")

frame_count = max(2, int(np.ceil(duration * VIDEO_FPS)) + 1)
frame_times = np.linspace(0.0, duration, frame_count)

frame_dist = np.interp(frame_times, elapsed, dist)
frame_vel = np.interp(frame_times, elapsed, vel)
frame_x = np.interp(frame_dist, dist, path_x)
frame_y = np.interp(frame_dist, dist, path_y)
frame_z = np.interp(frame_dist, dist, path_z)

# Exact endpoints: the tricycle starts and finishes on the route endpoints.
frame_x[0], frame_y[0], frame_z[0] = path_x[0], path_y[0], path_z[0]
frame_x[-1], frame_y[-1], frame_z[-1] = path_x[-1], path_y[-1], path_z[-1]

heading_x = np.gradient(frame_x)
heading_y = np.gradient(frame_y)
heading_norm = np.hypot(heading_x, heading_y)
heading_norm[heading_norm < 1e-9] = 1.0
heading_x /= heading_norm
heading_y /= heading_norm

finite_xy = np.isfinite(tx) & np.isfinite(ty)
terrain_xmin = float(np.min(tx[finite_xy]))
terrain_xmax = float(np.max(tx[finite_xy]))
terrain_ymin = float(np.min(ty[finite_xy]))
terrain_ymax = float(np.max(ty[finite_xy]))

finite_z = tz[np.isfinite(tz)]
if len(finite_z) == 0:
    raise ValueError("The terrain has no finite elevation values.")

terrain_zmin = float(np.min(finite_z))
terrain_zmax = float(np.max(finite_z))

print(f"Terrain array: {tz.shape}")
print(f"Route length: {route_length:,.1f} m")
print(f"Simulation duration: {duration:,.1f} s")
print(f"Frames: {frame_count:,}")
print(f"Start: ({frame_x[0]:.1f}, {frame_y[0]:.1f}, {frame_z[0]:.1f})")
print(f"Finish: ({frame_x[-1]:.1f}, {frame_y[-1]:.1f}, {frame_z[-1]:.1f})")
print(f"Rendering with up to {MAX_WORKERS} CPU cores.")

# Elevation band boundaries and colours, shared across all worker processes.
band_edges = [ELEVATION_BANDS[0][0]] + [
    band[1] for band in ELEVATION_BANDS
]
band_colours = [band[2] for band in ELEVATION_BANDS]
band_labels = [band[3] for band in ELEVATION_BANDS]
elevation_cmap = ListedColormap(band_colours, name="castle_hill_elevation")
elevation_norm = BoundaryNorm(band_edges, elevation_cmap.N, clip=True)

_PARALLEL_DATA = {
    "terrain_x": tx[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "terrain_y": ty[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "terrain_z": tz[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "road_x": rx,
    "road_y": ry,
    "road_z": rz,
    "times": frame_times,
    "distance": frame_dist,
    "velocity": frame_vel,
    "x": frame_x,
    "y": frame_y,
    "z": frame_z,
    "hx": heading_x,
    "hy": heading_y,
    "frame_count": frame_count,
    "terrain_xmin": terrain_xmin,
    "terrain_xmax": terrain_xmax,
    "terrain_ymin": terrain_ymin,
    "terrain_ymax": terrain_ymax,
    "terrain_zmin": terrain_zmin,
    "terrain_zmax": terrain_zmax,
    "camera_width": CAMERA_WIDTH,
    "camera_depth": CAMERA_DEPTH,
    "look_ahead": CAMERA_LOOK_AHEAD,
    "camera_elevation": CAMERA_ELEVATION,
    "turn_smoothing": CAMERA_TURN_SMOOTHING,
    "local_z_below": LOCAL_Z_BELOW,
    "local_z_above": LOCAL_Z_ABOVE,
    "vertical_aspect_factor": VERTICAL_ASPECT_FACTOR,
    "wheel_radius": WHEEL_RADIUS,
    "rear_track": REAR_TRACK,
    "rear_offset": REAR_OFFSET,
    "front_offset": FRONT_OFFSET,
    "video_dpi": VIDEO_DPI,
    "band_edges": band_edges,
    "band_colours": band_colours,
    "band_labels": band_labels,
}

_WORKER_SCENE = None

# ---------------------------------------------------------------------
# Initialize a scene in each worker
# ---------------------------------------------------------------------

def _initialize_render_worker():
    global _WORKER_SCENE

    d = _PARALLEL_DATA
    fig = plt.figure(figsize=(13.0, 9.0), dpi=d["video_dpi"])
    ax = fig.add_subplot(111, projection="3d")

    zz = d["terrain_z"]
    colours = ListedColormap(d["band_colours"])
    norm = BoundaryNorm(d["band_edges"], colours.N, clip=True)

    face_colours = colours(norm(zz))
    face_colours[..., 3] = np.where(np.isfinite(zz), 1.0, 0.0)

    ax.plot_surface(
        d["terrain_x"], d["terrain_y"], zz,
        facecolors=face_colours,
        linewidth=0,
        antialiased=False,
        shade=False,
    )

    # Draw the route over the terrain.
    ax.plot(
        d["road_x"], d["road_y"], d["road_z"] + 0.8,
        color="#181818", linewidth=3.2, zorder=15,
    )
    ax.plot(
        d["road_x"], d["road_y"], d["road_z"] + 1.6,
        color="white", linewidth=1.1, zorder=16,
    )

    trail, = ax.plot(
        [], [], [], color="#ff00cc", linewidth=3.0, zorder=25
    )

    names = [
        "rear_axle", "front_fork", "main_frame", "seat_stay",
        "handlebar", "crank", "seat", "torso", "arms", "legs",
    ]
    segments = {}
    for name in names:
        colour = "#ff8c00"
        width = 4.5
        if name in ("torso", "arms", "legs"):
            colour = "#b21f24"
            width = 5.5
        line, = ax.plot(
            [], [], [], color=colour, linewidth=width,
            solid_capstyle="round", zorder=40,
        )
        segments[name] = line

    wheels, hubs = [], []
    for _ in range(3):
        wheel, = ax.plot(
            [], [], [], color="#111111", linewidth=3.5, zorder=42
        )
        wheels.append(wheel)
        hub, = ax.plot(
            [], [], [], marker="o", markersize=5,
            linestyle="None", color="white",
            markeredgecolor="black", markeredgewidth=1.2, zorder=44,
        )
        hubs.append(hub)

    head, = ax.plot(
        [], [], [], marker="o", markersize=11,
        linestyle="None", markerfacecolor="#ffd28a",
        markeredgecolor="black", markeredgewidth=1.5, zorder=48,
    )

    fig.text(
        0.5, 0.975, "Castle Hill Road descent",
        ha="center", va="top", fontsize=16, fontweight="bold",
    )
    status = fig.text(
        0.5, 0.935, "", ha="center", va="top", fontsize=11,
    )

    # Fixed legend for the elevation bands.
    legend_handles = [
        Patch(facecolor=colour, edgecolor="#333333", label=label)
        for colour, label in zip(d["band_colours"], d["band_labels"])
    ]
    fig.legend(
        handles=legend_handles,
        title="Elevation above datum",
        loc="lower center",
        bbox_to_anchor=(0.5, 0.015),
        ncol=6,
        frameon=True,
        fontsize=9,
        title_fontsize=10,
    )

    # Remove all axes furniture; retain the 3D projection and terrain.
    ax.set_axis_off()
    ax.grid(False)
    ax.set_facecolor("#eeeeee")
    ax.set_proj_type("persp", focal_length=0.9)
    fig.subplots_adjust(left=0.01, right=0.99, bottom=0.10, top=0.90)

    _WORKER_SCENE = {
        "fig": fig,
        "ax": ax,
        "trail": trail,
        "segments": segments,
        "wheels": wheels,
        "hubs": hubs,
        "head": head,
        "status": status,
        "last_azimuth": None,
    }

# ---------------------------------------------------------------------
# Render one frame
# ---------------------------------------------------------------------

def _render_one_frame(i, output_directory):
    global _WORKER_SCENE

    if _WORKER_SCENE is None:
        _initialize_render_worker()

    d = _PARALLEL_DATA
    scene = _WORKER_SCENE
    ax = scene["ax"]

    x, y, z = float(d["x"][i]), float(d["y"][i]), float(d["z"][i])

    forward = np.array([d["hx"][i], d["hy"][i]], dtype=float)
    n = np.linalg.norm(forward)
    forward = forward / n if n > 1e-9 else np.array([1.0, 0.0])
    sideways = np.array([-forward[1], forward[0]])

    def p(longitudinal, lateral=0.0, height=0.0):
        return np.array([
            x + longitudinal * forward[0] + lateral * sideways[0],
            y + longitudinal * forward[1] + lateral * sideways[1],
            z + height,
        ])

    r = d["wheel_radius"]
    track = d["rear_track"]
    rear_left = p(d["rear_offset"], -track / 2, r)
    rear_right = p(d["rear_offset"], track / 2, r)
    front = p(d["front_offset"], 0.0, r)
    rear_mid = (rear_left + rear_right) / 2.0

    theta = np.linspace(0, 2 * np.pi, 48)
    for wheel, hub, centre in zip(
        scene["wheels"], scene["hubs"], [rear_left, rear_right, front]
    ):
        wheel.set_data_3d(
            centre[0] + r * np.cos(theta) * forward[0],
            centre[1] + r * np.cos(theta) * forward[1],
            centre[2] + r * np.sin(theta),
        )
        hub.set_data_3d([centre[0]], [centre[1]], [centre[2]])

    crank = p(-0.05 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE)
    seat = p(-0.42 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE)
    head_tube = p(0.40 * TRIKE_SCALE, 0, 0.95 * TRIKE_SCALE)
    handlebar = p(0.48 * TRIKE_SCALE, 0, 1.10 * TRIKE_SCALE)

    geometry = {
        "rear_axle": [rear_left, rear_right],
        "front_fork": [head_tube, front],
        "main_frame": [rear_mid, crank, head_tube],
        "seat_stay": [rear_mid, seat, crank],
        "handlebar": [head_tube, handlebar],
        "crank": [
            p(-0.18 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE),
            p(0.08 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE),
        ],
        "seat": [
            p(-0.55 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE),
            p(-0.30 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE),
        ],
    }

    hip = p(-0.38 * TRIKE_SCALE, 0, 1.30 * TRIKE_SCALE)
    shoulder = p(-0.10 * TRIKE_SCALE, 0, 1.80 * TRIKE_SCALE)
    head_centre = p(-0.08 * TRIKE_SCALE, 0, 2.02 * TRIKE_SCALE)
    hand = p(0.42 * TRIKE_SCALE, 0, 1.12 * TRIKE_SCALE)
    knee = p(-0.05 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE)
    foot = p(0.02 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE)

    geometry["torso"] = [hip, shoulder]
    geometry["arms"] = [shoulder, hand]
    geometry["legs"] = [hip, knee, foot]

    for name, points in geometry.items():
        arr = np.asarray(points)
        scene["segments"][name].set_data_3d(
            arr[:, 0], arr[:, 1], arr[:, 2]
        )

    scene["head"].set_data_3d(
        [head_centre[0]], [head_centre[1]], [head_centre[2]]
    )

    trail_start = max(0, i - VIDEO_FPS * 8)
    scene["trail"].set_data_3d(
        d["x"][trail_start:i + 1],
        d["y"][trail_start:i + 1],
        d["z"][trail_start:i + 1] + 1.5,
    )

    # Smoothly rotate the camera along the direction of travel.
    target_azimuth = np.degrees(np.arctan2(forward[1], forward[0])) + 180.0
    previous = scene["last_azimuth"]

    if previous is None:
        azimuth = target_azimuth
    else:
        delta = (target_azimuth - previous + 180.0) % 360.0 - 180.0
        azimuth = previous + d["turn_smoothing"] * delta

    scene["last_azimuth"] = azimuth
    ax.view_init(elev=d["camera_elevation"], azim=azimuth)

    # Track the tricycle with a close-up camera. Keep the view within
    # the terrain's rectangular extent.
    cx = x + forward[0] * d["look_ahead"]
    cy = y + forward[1] * d["look_ahead"]
    half_w = d["camera_width"] / 2.0
    half_h = d["camera_depth"] / 2.0

    x0, x1 = cx - half_w, cx + half_w
    y0, y1 = cy - half_h, cy + half_h

    if x0 < d["terrain_xmin"]:
        x1 += d["terrain_xmin"] - x0
        x0 = d["terrain_xmin"]
    if x1 > d["terrain_xmax"]:
        x0 -= x1 - d["terrain_xmax"]
        x1 = d["terrain_xmax"]
    if y0 < d["terrain_ymin"]:
        y1 += d["terrain_ymin"] - y0
        y0 = d["terrain_ymin"]
    if y1 > d["terrain_ymax"]:
        y0 -= y1 - d["terrain_ymax"]
        y1 = d["terrain_ymax"]

    x0 = max(x0, d["terrain_xmin"])
    x1 = min(x1, d["terrain_xmax"])
    y0 = max(y0, d["terrain_ymin"])
    y1 = min(y1, d["terrain_ymax"])

    ax.set_xlim(x0, x1)
    ax.set_ylim(y0, y1)

    z0 = max(d["terrain_zmin"], z - d["local_z_below"])
    z1 = min(d["terrain_zmax"], z + d["local_z_above"])
    if z1 - z0 < 35.0:
        z0 = max(d["terrain_zmin"], z - 35.0)
        z1 = min(d["terrain_zmax"], z + 35.0)
    if z1 <= z0:
        z0, z1 = d["terrain_zmin"], d["terrain_zmax"]

    ax.set_zlim(z0, z1)
    ax.set_box_aspect((
        max(x1 - x0, 1.0),
        max(y1 - y0, 1.0),
        max((z1 - z0) * d["vertical_aspect_factor"], 1.0),
    ))

    # Find the current terrain elevation from the nearest DEM grid cell.
    # The status label is informational; the vehicle position still uses
    # the route's interpolated elevation.
    terrain_elevation = z

    status_text = (
        f"Simulation time {d['times'][i]:,.1f} s"
        f"   |   Velocity {d['velocity'][i]:,.2f} m/s"
        f" ({d['velocity'][i] * 3.6:,.1f} km/h)"
        f"   |   Distance {d['distance'][i]:,.0f} m"
        f"   |   Elevation {terrain_elevation:,.1f} m"
    )
    scene["status"].set_text(status_text)

    frame_path = os.path.join(output_directory, f"frame_{i:06d}.jpg")
    scene["fig"].savefig(
        frame_path,
        dpi=d["video_dpi"],
        format="jpg",
        pil_kwargs={"quality": 92},
    )

    return int(i)

# ---------------------------------------------------------------------
# Render in parallel, report progress, and encode the MP4
# ---------------------------------------------------------------------

ffmpeg_path = shutil.which("ffmpeg")
if ffmpeg_path is None:
    raise RuntimeError("FFmpeg was not found. Install it and ensure ffmpeg is on PATH.")
if "fork" not in mp.get_all_start_methods():
    raise RuntimeError("This cell requires Linux multiprocessing with the fork start method.")

with tempfile.TemporaryDirectory(prefix="castle_hill_frames_") as frame_dir:
    context = mp.get_context("fork")
    completed = 0
    last_line_length = 0

    with ProcessPoolExecutor(
        max_workers=MAX_WORKERS,
        mp_context=context,
        initializer=_initialize_render_worker,
    ) as executor:
        futures = {
            executor.submit(_render_one_frame, i, frame_dir): i
            for i in range(frame_count)
        }

        for future in as_completed(futures):
            finished_frame = future.result()
            completed += 1
            percentage = 100.0 * completed / frame_count

            message = (
                f"Completed frame {finished_frame + 1:,}/{frame_count:,}"
                f" | Overall progress: {percentage:6.2f}%"
            )
            print(
                "\r" + message + " " * max(0, last_line_length - len(message)),
                end="",
                flush=True,
            )
            last_line_length = len(message)

    print("\nAll frames rendered. Encoding MP4...")

    subprocess.run(
        [
            ffmpeg_path,
            "-y",
            "-hide_banner",
            "-loglevel", "warning",
            "-framerate", str(VIDEO_FPS),
            "-i", os.path.join(frame_dir, "frame_%06d.jpg"),
            "-c:v", "libx264",
            "-preset", "medium",
            "-crf", str(VIDEO_CRF),
            "-pix_fmt", "yuv420p",
            "-threads", str(MAX_WORKERS),
            "-movflags", "+faststart",
            OUTPUT_FILE,
        ],
        check=True,
    )

print(f"\nFinished: {os.path.abspath(OUTPUT_FILE)}")
print(f"File size: {os.path.getsize(OUTPUT_FILE) / (1024 ** 2):,.1f} MiB")
```

```python

import os
import shutil
import tempfile
import subprocess
from pathlib import Path
from concurrent.futures import ProcessPoolExecutor, as_completed
import multiprocessing as mp

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FFMpegWriter
from IPython.display import Video, display

# ============================================================
# Configuration
# ============================================================

Crr_animation = 0.006
CdA_animation = 0.8

X_LIMITS = (-1500.0, 1000.0)
Y_LIMITS = (-800.0, 1000.0)

VERTICAL_EXAGGERATION = 4.0
TERRAIN_STRIDE = 3
VIDEO_FPS = 10
WORKERS = min(8, os.cpu_count() or 1)
DPI = 90

OUTPUT_PATH = Path.cwd() / "castle_hill_terrain_descent.mp4"

# ============================================================
# Check prerequisites and existing profile/terrain arrays
# ============================================================

if shutil.which("ffmpeg") is None:
    raise RuntimeError(
        "FFmpeg is not installed or is not on PATH."
    )

required = [
    "terrain_x", "terrain_y", "terrain_z",
    "road_x", "road_y", "road_z",
    "distance", "velocity",
]

missing = [name for name in required if name not in globals()]
if missing:
    raise RuntimeError(
        "Missing variables from the earlier terrain/profile cell: "
        + ", ".join(missing)
        + ". Rerun that cell first."
    )

terrain_x = np.asarray(terrain_x, dtype=float)
terrain_y = np.asarray(terrain_y, dtype=float)
terrain_z = np.asarray(terrain_z, dtype=float)

road_x = np.asarray(road_x, dtype=float)
road_y = np.asarray(road_y, dtype=float)
road_z = np.asarray(road_z, dtype=float)

distance = np.asarray(distance, dtype=float)
velocity = np.asarray(velocity, dtype=float)

if not (terrain_x.shape == terrain_y.shape == terrain_z.shape):
    raise ValueError("Terrain coordinate arrays have inconsistent shapes.")

if not (
    len(distance) == len(velocity)
    == len(road_x) == len(road_y) == len(road_z)
):
    raise ValueError("Route, distance and velocity arrays have inconsistent lengths.")

if len(distance) < 2 or np.any(np.diff(distance) <= 0):
    raise ValueError("Distance must be strictly increasing.")

if not all(
    np.all(np.isfinite(a))
    for a in (distance, velocity, road_x, road_y, road_z)
):
    raise ValueError("Route/profile arrays contain non-finite values.")

# ============================================================
# Construct the simulation time axis
# ============================================================

# Integrate dt = ds/v using trapezoidal integration of reciprocal
# velocity, consistent with the sampled velocity profile.
positive_velocity = np.maximum(velocity, 0.25)

segment_time = (
    0.5
    * (
        1.0 / positive_velocity[:-1]
        + 1.0 / positive_velocity[1:]
    )
    * np.diff(distance)
)

elapsed_time = np.concatenate(([0.0], np.cumsum(segment_time)))
total_time = float(elapsed_time[-1])

if not np.isfinite(total_time) or total_time <= 0:
    raise ValueError("Calculated simulation duration is invalid.")

# One frame per VIDEO_FPS fraction of a simulated second.
# Include the final frame at the end of the simulation.
frame_count = max(2, int(np.ceil(total_time * VIDEO_FPS)) + 1)
frame_times = np.linspace(0.0, total_time, frame_count)

frame_distance = np.interp(
    frame_times, elapsed_time, distance
)
frame_x = np.interp(frame_distance, distance, road_x)
frame_y = np.interp(frame_distance, distance, road_y)
frame_z = np.interp(frame_distance, distance, road_z)
frame_velocity = np.interp(frame_distance, distance, velocity)

print(f"Simulation duration: {total_time:.1f} seconds")
print(f"Video frames: {frame_count:,} at {VIDEO_FPS} fps")
print(f"Parallel workers: {WORKERS}")
print(f"Output: {OUTPUT_PATH}")

# ============================================================
# Crop the terrain to the requested display limits
# ============================================================

row_inside = (
    (terrain_y >= Y_LIMITS[0])
    & (terrain_y <= Y_LIMITS[1])
)
col_inside = (
    (terrain_x >= X_LIMITS[0])
    & (terrain_x <= X_LIMITS[1])
)

rows_to_keep = np.flatnonzero(np.any(row_inside, axis=1))
cols_to_keep = np.flatnonzero(np.any(col_inside, axis=0))

if len(rows_to_keep) == 0 or len(cols_to_keep) == 0:
    raise ValueError("Terrain does not intersect the requested display bounds.")

r0, r1 = rows_to_keep[0], rows_to_keep[-1] + 1
c0, c1 = cols_to_keep[0], cols_to_keep[-1] + 1

tx = terrain_x[r0:r1, c0:c1].copy()
ty = terrain_y[r0:r1, c0:c1].copy()
tz = terrain_z[r0:r1, c0:c1].copy()

inside = (
    (tx >= X_LIMITS[0])
    & (tx <= X_LIMITS[1])
    & (ty >= Y_LIMITS[0])
    & (ty <= Y_LIMITS[1])
)

tz[~inside] = np.nan

valid_elevations = tz[np.isfinite(tz)]
if valid_elevations.size == 0:
    raise ValueError("No valid terrain elevations inside the requested bounds.")

z_min = float(valid_elevations.min())
z_max = float(valid_elevations.max())
z_padding = max(5.0, (z_max - z_min) * 0.04)
Z_LIMITS = (z_min - z_padding, z_max + z_padding)

print(f"Cropped terrain grid: {tz.shape[1]} x {tz.shape[0]}")
print(f"Elevation limits: {Z_LIMITS[0]:.1f}–{Z_LIMITS[1]:.1f} m")

# ============================================================
# Share scene data with workers using Linux fork
# ============================================================

# Linux fork allows workers to inherit these read-only arrays
# without serializing and copying the complete DEM for every task.
_PARALLEL_DATA = {
    "tx": tx,
    "ty": ty,
    "tz": tz,
    "road_x": road_x,
    "road_y": road_y,
    "road_z": road_z,
    "frame_x": frame_x,
    "frame_y": frame_y,
    "frame_z": frame_z,
    "frame_velocity": frame_velocity,
    "frame_distance": frame_distance,
    "frame_times": frame_times,
    "total_time": total_time,
    "frame_count": frame_count,
    "Crr": Crr_animation,
    "CdA": CdA_animation,
    "x_limits": X_LIMITS,
    "y_limits": Y_LIMITS,
    "z_limits": Z_LIMITS,
    "terrain_stride": TERRAIN_STRIDE,
    "vertical_exaggeration": VERTICAL_EXAGGERATION,
    "dpi": DPI,
}

_WORKER_SCENE = None

def _initialize_render_worker():
    """Create an independent Matplotlib scene in each worker."""
    global _WORKER_SCENE

    import matplotlib
    matplotlib.use("Agg", force=True)
    import matplotlib.pyplot as worker_plt

    d = _PARALLEL_DATA

    fig = worker_plt.figure(figsize=(12, 9), dpi=d["dpi"])
    ax = fig.add_axes(
        [0.035, 0.065, 0.835, 0.82],
        projection="3d",
    )
    cax = fig.add_axes([0.895, 0.24, 0.018, 0.48])

    stride = d["terrain_stride"]

    surface = ax.plot_surface(
        d["tx"][::stride, ::stride],
        d["ty"][::stride, ::stride],
        d["tz"][::stride, ::stride],
        cmap="terrain",
        linewidth=0,
        antialiased=True,
        shade=True,
        alpha=0.97,
    )

    colorbar = fig.colorbar(surface, cax=cax)
    colorbar.set_label("Elevation (m)")

    ax.plot(
        d["road_x"],
        d["road_y"],
        d["road_z"] + 1.5,
        color="red",
        linewidth=2.2,
        label="Road",
    )

    trail, = ax.plot(
        [], [], [],
        color="deepskyblue",
        linewidth=3.0,
        label="Cyclist's path",
    )

    marker, = ax.plot(
        [], [], [],
        marker="o",
        markersize=9,
        markerfacecolor="blue",
        markeredgecolor="white",
        markeredgewidth=1.2,
        linestyle="None",
        label="Cyclist",
    )

    ax.set_xlim(*d["x_limits"])
    ax.set_ylim(*d["y_limits"])
    ax.set_zlim(*d["z_limits"])

    ax.set_xlabel("West–east distance (m)", labelpad=9)
    ax.set_ylabel("South–north distance (m)", labelpad=9)
    ax.set_zlabel("Elevation (m)", labelpad=8)

    ax.set_box_aspect((
        d["x_limits"][1] - d["x_limits"][0],
        d["y_limits"][1] - d["y_limits"][0],
        (d["z_limits"][1] - d["z_limits"][0])
        * d["vertical_exaggeration"],
    ))

    ax.set_proj_type("ortho")
    ax.view_init(elev=38, azim=-60)
    ax.tick_params(axis="both", labelsize=8)
    ax.tick_params(axis="z", labelsize=8)
    ax.legend(loc="upper left", fontsize=9)

    fig.text(
        0.48, 0.975,
        "Castle Hill Road descent — 3D terrain",
        ha="center",
        va="top",
        fontsize=16,
        fontweight="bold",
    )

    status = fig.text(
        0.48, 0.942, "",
        ha="center",
        va="top",
        fontsize=10,
    )

    _WORKER_SCENE = {
        "fig": fig,
        "ax": ax,
        "trail": trail,
        "marker": marker,
        "status": status,
    }


def _render_one_frame(frame_index, output_filename):
    """Render one frame to a JPEG file."""
    import matplotlib.pyplot as worker_plt

    d = _PARALLEL_DATA
    scene = _WORKER_SCENE

    i = frame_index
    t = float(d["frame_times"][i])

    scene["trail"].set_data_3d(
        d["frame_x"][:i + 1],
        d["frame_y"][:i + 1],
        d["frame_z"][:i + 1] + 1.5,
    )

    scene["marker"].set_data_3d(
        [d["frame_x"][i]],
        [d["frame_y"][i]],
        [d["frame_z"][i] + 3.0],
    )

    hours = int(t // 3600)
    minutes = int((t % 3600) // 60)
    seconds = int(t % 60)

    scene["status"].set_text(
        f"Crr = {d['Crr']:.3f}   |   "
        f"CdA = {d['CdA']:.1f} m²   |   "
        f"Velocity = {d['frame_velocity'][i] * 3.6:.1f} km/h   |   "
        f"Distance = {d['frame_distance'][i]:.0f} / "
        f"{d['frame_distance'][-1]:.0f} m   |   "
        f"Simulation time = {hours:02d}:{minutes:02d}:{seconds:02d}"
    )

    filename = Path(output_filename)

    # JPEG frames reduce temporary disk usage; final output is MP4.
    scene["fig"].savefig(
        filename,
        format="jpeg",
        dpi=d["dpi"],
        pil_kwargs={"quality": 95},
    )

    return i


# ============================================================
# Render frames in parallel, then encode the MP4
# ============================================================

if "fork" not in mp.get_all_start_methods():
    raise RuntimeError(
        "This parallel-rendering cell requires Linux's fork "
        "multiprocessing start method."
    )

temporary_directory = tempfile.TemporaryDirectory(
    prefix="castle_hill_frames_"
)
frames_dir = Path(temporary_directory.name)

try:
    context = mp.get_context("fork")

    print("Rendering frames in parallel...")

    completed = 0

    with ProcessPoolExecutor(
        max_workers=WORKERS,
        mp_context=context,
        initializer=_initialize_render_worker,
    ) as executor:

        futures = [
            executor.submit(
                _render_one_frame,
                i,
                str(frames_dir / f"frame_{i:06d}.jpg"),
            )
            for i in range(frame_count)
        ]

        for future in as_completed(futures):
            future.result()
            completed += 1

            if completed % 10 == 0 or completed == frame_count:
                print(
                    f"\rRendered {completed:,}/{frame_count:,} frames "
                    f"({100 * completed / frame_count:.1f}%)",
                    end="",
                    flush=True,
                )

    print("\nEncoding MP4 with FFmpeg...")

    command = [
        shutil.which("ffmpeg"),
        "-hide_banner",
        "-loglevel", "error",
        "-y",
        "-framerate", str(VIDEO_FPS),
        "-start_number", "0",
        "-i", str(frames_dir / "frame_%06d.jpg"),
        "-frames:v", str(frame_count),
        "-c:v", "libx264",
        "-preset", "medium",
        "-crf", "18",
        "-pix_fmt", "yuv420p",
        "-threads", str(WORKERS),
        "-movflags", "+faststart",
        str(OUTPUT_PATH),
    ]

    result = subprocess.run(
        command,
        text=True,
        capture_output=True,
    )

    if result.returncode != 0:
        raise RuntimeError(
            "FFmpeg failed to encode the video:\n"
            + result.stderr[-5000:]
        )

    if not OUTPUT_PATH.is_file() or OUTPUT_PATH.stat().st_size == 0:
        raise RuntimeError("The MP4 file was not created successfully.")

    print(f"Created: {OUTPUT_PATH.resolve()}")
    print(f"File size: {OUTPUT_PATH.stat().st_size / (1024**2):.2f} MiB")
    print(f"Duration represented by simulation: {total_time:.2f} seconds")

finally:
    # The temporary frame directory is deleted automatically.
    temporary_directory.cleanup()

display(Video(str(OUTPUT_PATH), embed=False, width=1100))
```

```python
# COMPLETE SINGLE-CELL WORKFLOW: fetch the route and QldDem terrain,
# calculate a downhill speed profile, then render the smoothed-camera animation.
# Requires: numpy, scipy, requests, matplotlib and ffmpeg.
# The route/terrain queries require an internet connection.

import json
import math
import requests
import numpy as np
from scipy.interpolate import PchipInterpolator
from scipy.integrate import solve_ivp
from scipy.signal import savgol_filter

print("Fetching Castle Hill Road route from OpenStreetMap/OSRM...")

def _geocode(address):
    response = requests.get(
        "https://nominatim.openstreetmap.org/search",
        params={"q": address, "format": "json", "countrycodes": "au", "limit": 10},
        headers={"User-Agent": "Castle-Hill-terrain-animation/1.0"},
        timeout=30,
    )
    response.raise_for_status()
    results = response.json()
    if not results:
        raise RuntimeError(f"No geocoding result for {address!r}")
    for item in results:
        if "stanley street" in item.get("display_name", "").lower():
            return float(item["lon"]), float(item["lat"])
    raise RuntimeError(f"Could not identify Stanley Street in geocoding results: {address!r}")

start_lon, start_lat = _geocode(
    "320 Stanley Street, North Ward, Queensland, Australia"
)
summit_lon, summit_lat = 146.8026, -19.2580

# Route uphill from Stanley Street to the summit, then reverse it for descent.
_route_url = (
    "https://router.project-osrm.org/route/v1/driving/"
    f"{start_lon},{start_lat};{summit_lon},{summit_lat}"
)
_route_response = requests.get(
    _route_url,
    params={"overview": "full", "geometries": "geojson", "steps": "true"},
    timeout=60,
)
_route_response.raise_for_status()
_route_result = _route_response.json()
if _route_result.get("code") != "Ok":
    raise RuntimeError(f"OSRM routing failed: {_route_result}")
_route = _route_result["routes"][0]
_route_coords = np.asarray(_route["geometry"]["coordinates"], dtype=float)
_route_names = [
    step.get("name", "").strip()
    for step in _route.get("legs", [{}])[0].get("steps", [])
]
for _street in ("Gregory Street", "Stanley Street", "Castle Hill Road"):
    if not any(_street.lower() in name.lower() for name in _route_names):
        raise RuntimeError(
            f"OSRM route did not include {_street!r}; route steps were {_route_names!r}"
        )

# Convert lon/lat to local metric coordinates centred on Castle Hill.
_center_lon = float(np.mean(_route_coords[:, 0]))
_center_lat = float(np.mean(_route_coords[:, 1]))
_m_per_deg_x = 111320.0 * math.cos(math.radians(_center_lat))
_m_per_deg_y = 110574.0
def _local_xy(lon, lat):
    return (
        (np.asarray(lon, dtype=float) - _center_lon) * _m_per_deg_x,
        (np.asarray(lat, dtype=float) - _center_lat) * _m_per_deg_y,
    )

# Query QldDem elevation samples in bounded chunks.
_dem_url = (
    "https://spatial-img.information.qld.gov.au/"
    "arcgis/rest/services/Elevation/QldDem/ImageServer/getSamples"
)



def _query_qlddem(lons, lats, chunk_size=100, max_retries=3):
    import time

    lons = np.asarray(lons, dtype=float).ravel()
    lats = np.asarray(lats, dtype=float).ravel()

    if len(lons) != len(lats):
        raise ValueError(
            f"Longitude and latitude arrays differ in length: "
            f"{len(lons)} != {len(lats)}"
        )
    if chunk_size < 1:
        raise ValueError("chunk_size must be at least 1")
    if max_retries < 0:
        raise ValueError("max_retries cannot be negative")

    values = np.full(len(lons), np.nan, dtype=float)
    total = len(lons)

    def _fetch_batch(begin, end):
        geometry = {
            "points": [
                [float(lon), float(lat)]
                for lon, lat in zip(lons[begin:end], lats[begin:end])
            ],
            "spatialReference": {"wkid": 4326},
        }
        params = {
            "geometry": json.dumps(geometry),
            "geometryType": "esriGeometryMultipoint",
            "returnGeometry": "false",
            "returnCatalogItems": "false",
            "f": "json",
        }

        last_error = None

        for attempt in range(max_retries + 1):
            try:
                response = requests.get(
                    _identify_url,
                    params=params,
                    timeout=60,
                )
                response.raise_for_status()
                payload = response.json()

                if "error" in payload:
                    raise RuntimeError(
                        f"QldDem identify error: {payload['error']}"
                    )

                results = payload.get("results")
                if not isinstance(results, list):
                    raise RuntimeError(
                        f"Unexpected QldDem identify response: {payload}"
                    )

                # Identify may omit points with no returned raster value.
                # Do not assume a shorter response aligns positionally.
                if len(results) != end - begin:
                    raise RuntimeError(
                        f"QldDem identify returned {len(results)} results "
                        f"for {end - begin} points"
                    )

                batch_values = np.full(end - begin, np.nan)

                for j, result in enumerate(results):
                    try:
                        batch_values[j] = float(result["value"])
                    except (KeyError, TypeError, ValueError):
                        pass

                if not np.all(np.isfinite(batch_values)):
                    raise RuntimeError(
                        "QldDem identify returned missing or non-finite "
                        "elevations"
                    )

                return batch_values

            except (
                requests.RequestException,
                ValueError,
                RuntimeError,
            ) as exc:
                last_error = exc

                if attempt < max_retries:
                    delay = min(2 ** attempt, 8)
                    print(
                        f"\nQldDem identify failed for samples "
                        f"{begin:,}–{end - 1:,} "
                        f"(attempt {attempt + 1}/{max_retries + 1}): "
                        f"{exc}. Retrying in {delay}s...",
                        flush=True,
                    )
                    time.sleep(delay)

        if end - begin > 1:
            midpoint = begin + (end - begin) // 2
            left = _fetch_batch(begin, midpoint)
            right = _fetch_batch(midpoint, end)
            return np.concatenate((left, right))

        raise RuntimeError(
            f"QldDem identify failed for sample {begin}: "
            f"longitude={lons[begin]:.8f}, "
            f"latitude={lats[begin]:.8f}. Last error: {last_error}"
        ) from last_error

    for begin in range(0, total, chunk_size):
        end = min(begin + chunk_size, total)
        values[begin:end] = _fetch_batch(begin, end)
        print(
            f"\rQldDem samples: {end:,}/{total:,}",
            end="",
            flush=True,
        )

    print()

    if not np.all(np.isfinite(values)):
        raise RuntimeError(
            f"QldDem returned "
            f"{np.count_nonzero(~np.isfinite(values))} invalid elevations"
        )

    return values



# Sample the route at roughly 10 m spacing and query its elevation profile.
def _haversine(lon1, lat1, lon2, lat2):
    r = 6371000.0
    p1, p2 = math.radians(lat1), math.radians(lat2)
    dp = math.radians(lat2-lat1)
    dl = math.radians(lon2-lon1)
    a = math.sin(dp/2)**2 + math.cos(p1)*math.cos(p2)*math.sin(dl/2)**2
    return 2*r*math.asin(math.sqrt(a))

_route_s = np.zeros(len(_route_coords), dtype=float)
for _i in range(1, len(_route_coords)):
    _route_s[_i] = _route_s[_i-1] + _haversine(
        *_route_coords[_i-1], *_route_coords[_i]
    )
if not 2000.0 <= _route_s[-1] <= 4000.0:
    raise RuntimeError(f"Unexpected Castle Hill route length: {_route_s[-1]:.1f} m")
_profile_s = np.arange(0.0, _route_s[-1], 10.0)
if len(_profile_s) == 0 or _profile_s[-1] < _route_s[-1]:
    _profile_s = np.append(_profile_s, _route_s[-1])
_route_lon = np.interp(_profile_s, _route_s, _route_coords[:, 0])
_route_lat = np.interp(_profile_s, _route_s, _route_coords[:, 1])
print(f"Route length: {_route_s[-1]/1000:.3f} km")

print("Querying QldDem for the route elevation profile...")
_profile_z = _query_qlddem(_route_lon, _route_lat, chunk_size=3)
if len(_profile_z) >= 11:
    _profile_z = savgol_filter(_profile_z, 11, 3)

# Orient descent from summit to Stanley Street.
distance = (_profile_s[-1] - _profile_s[::-1]).copy()
distance -= distance[0]
h = _profile_z[::-1].copy()
road_lon = _route_lon[::-1].copy()
road_lat = _route_lat[::-1].copy()
road_x, road_y = _local_xy(road_lon, road_lat)
road_z = h.copy()

# A modest-resolution terrain mesh around the route. The broad grid provides
# surrounding relief while keeping the number of DEM requests manageable.
_grid_spacing = 65.0
_margin = 1800.0
_xmin, _xmax = float(np.min(road_x)-_margin), float(np.max(road_x)+_margin)
_ymin, _ymax = float(np.min(road_y)-_margin), float(np.max(road_y)+_margin)
_grid_x = np.arange(_xmin, _xmax + _grid_spacing, _grid_spacing)
_grid_y = np.arange(_ymin, _ymax + _grid_spacing, _grid_spacing)
terrain_x, terrain_y = np.meshgrid(_grid_x, _grid_y)
_grid_lon = terrain_x.ravel() / _m_per_deg_x + _center_lon
_grid_lat = terrain_y.ravel() / _m_per_deg_y + _center_lat
print(
    f"Querying QldDem for terrain mesh ({terrain_x.shape[1]} x {terrain_x.shape[0]} points)..."
)
terrain_z = _query_qlddem(_grid_lon, _grid_lat, chunk_size=3).reshape(terrain_x.shape)

# Calculate one representative brake-failure/coasting speed profile.
# These assumptions match the physical model used in the earlier analysis.
mass = 64.0 + 20.0 + 5.0
gravity = 9.81
air_density = 1.225
rolling_resistance = 0.006
drag_area = 0.50
slope = np.gradient(h, distance)
sin_theta = np.clip(-slope, -1.0, 1.0)
cos_theta = np.sqrt(np.maximum(0.0, 1.0 - sin_theta**2))
sin_fn = PchipInterpolator(distance, sin_theta, extrapolate=True)
cos_fn = PchipInterpolator(distance, cos_theta, extrapolate=True)
def _speed_ode(s_pos, w):
    st, ct = float(sin_fn(s_pos)), float(cos_fn(s_pos))
    w0 = max(float(w[0]), 0.0)
    acc = gravity*st - rolling_resistance*gravity*ct - air_density*drag_area/(2*mass)*w0
    if w0 <= 0.0 and acc < 0.0:
        return [0.0]
    return [2.0*acc]
_solution = solve_ivp(
    _speed_ode, (float(distance[0]), float(distance[-1])), [0.0],
    t_eval=distance, rtol=1e-8, atol=1e-10, max_step=2.0,
)
if not _solution.success:
    raise RuntimeError(f"Speed integration failed: {_solution.message}")
velocity = np.sqrt(np.maximum(_solution.y[0], 0.0))

print(
    f"Prepared descent: {distance[-1]/1000:.3f} km; "
    f"elevation {h[0]:.1f} m -> {h[-1]:.1f} m; "
    f"peak speed {np.max(velocity)*3.6:.1f} km/h"
)
print("Starting the corrected animation render...")



# Castle Hill Road descent: close-up, terrain-following 3D animation.
#
# Features:
#   - Camera follows the tricycle and turns smoothly with the road.
#   - Zoomed-in view centred on the tricycle and nearby route.
#   - Dynamic elevation limits and an exaggerated vertical aspect.
#   - Full terrain and road retained in their original world coordinates.
#   - Enlarged schematic tricycle with three wheels, frame and rider.
#   - Elapsed simulation time, velocity and distance in the video.
#   - Live rendering progress: frame number and percentage.
#   - Parallel rendering across up to eight CPU cores.
#   - MP4 output.
#
# Required existing notebook variables:
#   terrain_x, terrain_y, terrain_z
#   road_x, road_y, road_z
#   distance, velocity

import os
import shutil
import subprocess
import tempfile
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor, as_completed

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib import cm

# ---------------------------------------------------------------------
# Configuration
# ---------------------------------------------------------------------

OUTPUT_FILE = "castle_hill_terrain_descent.mp4"

MAX_WORKERS = min(8, os.cpu_count() or 1)
VIDEO_FPS = 10
VIDEO_DPI = 100
VIDEO_CRF = 18
TERRAIN_STRIDE = 3

# Close-up camera window in metres.
CAMERA_WIDTH = 700.0
CAMERA_DEPTH = 550.0
CAMERA_LOOK_AHEAD = 100.0
CAMERA_ELEVATION = 38.0
CAMERA_TURN_SMOOTHING = 0.20

# Elevation window follows the local terrain.
LOCAL_Z_BELOW = 65.0
LOCAL_Z_ABOVE = 95.0

# Vertical visual exaggeration. Higher values make relief more prominent.
VERTICAL_ASPECT_FACTOR = 2.5

# Schematic tricycle dimensions are exaggerated to keep it visible.
TRIKE_SCALE = 4.0
WHEEL_RADIUS = 0.34 * TRIKE_SCALE
REAR_TRACK = 0.85 * TRIKE_SCALE
REAR_OFFSET = -0.85 * TRIKE_SCALE
FRONT_OFFSET = 0.90 * TRIKE_SCALE

# ---------------------------------------------------------------------
# Validate and prepare the notebook data
# ---------------------------------------------------------------------

required = (
    "terrain_x", "terrain_y", "terrain_z",
    "road_x", "road_y", "road_z",
    "distance", "velocity",
)
missing = [name for name in required if name not in globals()]
if missing:
    raise NameError("Missing notebook variables: " + ", ".join(missing))

tx = np.asarray(terrain_x, dtype=float)
ty = np.asarray(terrain_y, dtype=float)
tz = np.asarray(terrain_z, dtype=float)

rx = np.asarray(road_x, dtype=float).ravel()
ry = np.asarray(road_y, dtype=float).ravel()
rz = np.asarray(road_z, dtype=float).ravel()

dist = np.asarray(distance, dtype=float).ravel()
vel = np.asarray(velocity, dtype=float).ravel()

if tx.ndim != 2 or tx.shape != ty.shape or tx.shape != tz.shape:
    raise ValueError("terrain_x, terrain_y and terrain_z must have matching 2D shapes.")
if not (len(rx) == len(ry) == len(rz)):
    raise ValueError("road_x, road_y and road_z must have equal lengths.")
if len(rx) < 2 or len(dist) < 2 or len(dist) != len(vel):
    raise ValueError("The route and simulation arrays are incomplete.")

# Clean simulation samples and sort by increasing distance.
valid = np.isfinite(dist) & np.isfinite(vel)
dist, vel = dist[valid], vel[valid]
order = np.argsort(dist)
dist, vel = dist[order], vel[order]
dist, unique_idx = np.unique(dist, return_index=True)
vel = vel[unique_idx]

if len(dist) < 2 or dist[-1] <= dist[0]:
    raise ValueError("Simulation distance must be strictly increasing.")
if not (
    np.all(np.isfinite(rx))
    and np.all(np.isfinite(ry))
    and np.all(np.isfinite(rz))
):
    raise ValueError("The road arrays contain non-finite coordinates.")

# Orient the route using h, if available, as the summit-to-Gregory
# elevation reference. Otherwise use the higher endpoint as the start.
reverse_route = False
if "h" in globals():
    href = np.asarray(h, dtype=float).ravel()
    href = href[np.isfinite(href)]
    if len(href) >= 2:
        same_error = abs(rz[0] - href[0]) + abs(rz[-1] - href[-1])
        reverse_error = abs(rz[-1] - href[0]) + abs(rz[0] - href[-1])
        reverse_route = reverse_error < same_error
    else:
        reverse_route = rz[0] < rz[-1]
else:
    reverse_route = rz[0] < rz[-1]

if reverse_route:
    rx, ry, rz = rx[::-1].copy(), ry[::-1].copy(), rz[::-1].copy()
    print("Reversed road coordinates to match the descent direction.")

# Cumulative horizontal distance along the route.
route_ds = np.hypot(np.diff(rx), np.diff(ry))
route_s = np.concatenate(([0.0], np.cumsum(route_ds)))
route_s, route_unique_idx = np.unique(route_s, return_index=True)
rx, ry, rz = rx[route_unique_idx], ry[route_unique_idx], rz[route_unique_idx]

route_length = float(route_s[-1])
simulation_length = float(dist[-1] - dist[0])
if route_length <= 0 or simulation_length <= 0:
    raise ValueError("Could not determine a positive route or simulation length.")

# Map the simulation distance fraction directly to route distance.
# Both endpoints therefore correspond to the actual route endpoints.
route_progress = (dist - dist[0]) / simulation_length * route_length
path_x = np.interp(route_progress, route_s, rx)
path_y = np.interp(route_progress, route_s, ry)
path_z = np.interp(route_progress, route_s, rz)

# Integrate dt = ds/v to calculate elapsed simulation time.
safe_velocity = np.maximum(vel, 0.1)
elapsed = np.zeros_like(dist)
elapsed[1:] = np.cumsum(
    np.diff(dist) * 0.5 * (
        1.0 / safe_velocity[:-1] + 1.0 / safe_velocity[1:]
    )
)

duration = float(elapsed[-1])
if not np.isfinite(duration) or duration <= 0:
    raise ValueError("Could not calculate a valid simulation duration.")

frame_count = max(2, int(np.ceil(duration * VIDEO_FPS)) + 1)
frame_times = np.linspace(0.0, duration, frame_count)

frame_dist = np.interp(frame_times, elapsed, dist)
frame_vel = np.interp(frame_times, elapsed, vel)
frame_x = np.interp(frame_dist, dist, path_x)
frame_y = np.interp(frame_dist, dist, path_y)
frame_z = np.interp(frame_dist, dist, path_z)

# Guarantee exact endpoint positions.
frame_x[0], frame_y[0], frame_z[0] = path_x[0], path_y[0], path_z[0]
frame_x[-1], frame_y[-1], frame_z[-1] = path_x[-1], path_y[-1], path_z[-1]

# Direction of travel at every video frame.
heading_x = np.gradient(frame_x)
heading_y = np.gradient(frame_y)
heading_norm = np.hypot(heading_x, heading_y)
heading_norm[heading_norm < 1e-9] = 1.0
heading_x /= heading_norm
heading_y /= heading_norm

# Precompute camera rotation in chronological order before parallel rendering.
# Unwrap angles so turns across +/-180 degrees remain continuous, then smooth
# them sequentially. Workers must not infer camera state from frame completion.
raw_azimuth = np.unwrap(np.arctan2(heading_y, heading_x) + np.pi)

# A short centred moving average suppresses frame-to-frame direction noise.
heading_window = max(3, int(round(VIDEO_FPS * 0.8)))
if heading_window % 2 == 0:
    heading_window += 1
heading_kernel = np.ones(heading_window, dtype=float) / heading_window
pad = heading_window // 2
padded_azimuth = np.pad(raw_azimuth, (pad, pad), mode="edge")
smoothed_azimuth = np.convolve(padded_azimuth, heading_kernel, mode="valid")

# Apply the existing turn-smoothing setting in one deterministic pass.
camera_azimuth = np.empty_like(smoothed_azimuth)
camera_azimuth[0] = smoothed_azimuth[0]
for frame_i in range(1, len(camera_azimuth)):
    camera_azimuth[frame_i] = (
        camera_azimuth[frame_i - 1]
        + CAMERA_TURN_SMOOTHING
        * (smoothed_azimuth[frame_i] - camera_azimuth[frame_i - 1])
    )

# Terrain bounds, used to prevent the camera from looking outside the DEM.
finite_xy = np.isfinite(tx) & np.isfinite(ty)
terrain_xmin = float(np.min(tx[finite_xy]))
terrain_xmax = float(np.max(tx[finite_xy]))
terrain_ymin = float(np.min(ty[finite_xy]))
terrain_ymax = float(np.max(ty[finite_xy]))

finite_z = tz[np.isfinite(tz)]
if len(finite_z) == 0:
    raise ValueError("The terrain has no finite elevation values.")

terrain_zmin = float(np.percentile(finite_z, 1))
terrain_zmax = float(np.percentile(finite_z, 99))

print(f"Terrain array: {tz.shape}")
print(f"Route length: {route_length:,.1f} m")
print(f"Simulation duration: {duration:,.1f} s")
print(f"Frames: {frame_count:,}")
print(f"Start: ({frame_x[0]:.1f}, {frame_y[0]:.1f}, {frame_z[0]:.1f})")
print(f"Finish: ({frame_x[-1]:.1f}, {frame_y[-1]:.1f}, {frame_z[-1]:.1f})")
print(f"Rendering with up to {MAX_WORKERS} CPU cores.")

_PARALLEL_DATA = {
    "terrain_x": tx[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "terrain_y": ty[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "terrain_z": tz[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "road_x": rx,
    "road_y": ry,
    "road_z": rz,
    "times": frame_times,
    "distance": frame_dist,
    "velocity": frame_vel,
    "x": frame_x,
    "y": frame_y,
    "z": frame_z,
    "hx": heading_x,
    "hy": heading_y,
    "camera_azimuth": camera_azimuth,
    "frame_count": frame_count,
    "terrain_xmin": terrain_xmin,
    "terrain_xmax": terrain_xmax,
    "terrain_ymin": terrain_ymin,
    "terrain_ymax": terrain_ymax,
    "terrain_zmin": terrain_zmin,
    "terrain_zmax": terrain_zmax,
    "camera_width": CAMERA_WIDTH,
    "camera_depth": CAMERA_DEPTH,
    "look_ahead": CAMERA_LOOK_AHEAD,
    "camera_elevation": CAMERA_ELEVATION,
    "turn_smoothing": CAMERA_TURN_SMOOTHING,
    "local_z_below": LOCAL_Z_BELOW,
    "local_z_above": LOCAL_Z_ABOVE,
    "vertical_aspect_factor": VERTICAL_ASPECT_FACTOR,
    "wheel_radius": WHEEL_RADIUS,
    "rear_track": REAR_TRACK,
    "rear_offset": REAR_OFFSET,
    "front_offset": FRONT_OFFSET,
    "video_dpi": VIDEO_DPI,
}

_WORKER_SCENE = None

# ---------------------------------------------------------------------
# Initialize one scene in each worker
# ---------------------------------------------------------------------

def _initialize_render_worker():
    global _WORKER_SCENE

    d = _PARALLEL_DATA

    fig = plt.figure(figsize=(13.0, 9.0), dpi=d["video_dpi"])
    ax = fig.add_subplot(111, projection="3d")

    zz = d["terrain_z"]
    zmin = float(np.nanpercentile(zz, 2))
    zmax = float(np.nanpercentile(zz, 98))
    if zmax <= zmin:
        zmax = zmin + 1.0

    terrain_colours = cm.terrain(
        mpl.colors.Normalize(vmin=zmin, vmax=zmax)(zz)
    )
    terrain_colours[..., 3] = np.where(np.isfinite(zz), 1.0, 0.0)

    ax.plot_surface(
        d["terrain_x"], d["terrain_y"], zz,
        facecolors=terrain_colours,
        linewidth=0,
        antialiased=False,
        shade=True,
    )

    # Draw the route with a dark outline and bright centre.
    ax.plot(
        d["road_x"], d["road_y"], d["road_z"] + 0.8,
        color="black", linewidth=3.2, zorder=15,
    )
    ax.plot(
        d["road_x"], d["road_y"], d["road_z"] + 1.6,
        color="white", linewidth=1.1, zorder=16,
    )

    trail, = ax.plot(
        [], [], [], color="orangered", linewidth=3.0, zorder=25
    )

    names = [
        "rear_axle", "front_fork", "main_frame", "seat_stay",
        "handlebar", "crank", "seat", "torso", "arms", "legs",
    ]
    segments = {}
    for name in names:
        colour = "#ff8c00"
        width = 4.5
        if name in ("torso", "arms", "legs"):
            colour = "#b21f24"
            width = 5.5
        line, = ax.plot(
            [], [], [], color=colour, linewidth=width,
            solid_capstyle="round", zorder=40,
        )
        segments[name] = line

    wheels, hubs = [], []
    for _ in range(3):
        wheel, = ax.plot(
            [], [], [], color="#111111", linewidth=3.5, zorder=42
        )
        wheels.append(wheel)

        hub, = ax.plot(
            [], [], [], marker="o", markersize=5,
            linestyle="None", color="white",
            markeredgecolor="black", markeredgewidth=1.2, zorder=44,
        )
        hubs.append(hub)

    head, = ax.plot(
        [], [], [], marker="o", markersize=11,
        linestyle="None", markerfacecolor="#ffd28a",
        markeredgecolor="black", markeredgewidth=1.5, zorder=48,
    )

    fig.text(
        0.5, 0.975, "Castle Hill Road descent",
        ha="center", va="top", fontsize=16, fontweight="bold",
    )
    status = fig.text(
        0.5, 0.935, "", ha="center", va="top", fontsize=11,
    )

    # Reserve room for all three axis labels and the header.
    fig.subplots_adjust(left=0.04, right=0.88, bottom=0.12, top=0.89)

    ax.set_xlabel("West\u2013east position (m)", labelpad=10)
    ax.set_ylabel("South\u2013north position (m)", labelpad=10)
    ax.set_zlabel("Elevation (m)", labelpad=8)
    ax.set_facecolor("#eeeeee")
    ax.grid(True)
    ax.set_proj_type("persp", focal_length=0.9)

    _WORKER_SCENE = {
        "fig": fig,
        "ax": ax,
        "trail": trail,
        "segments": segments,
        "wheels": wheels,
        "hubs": hubs,
        "head": head,
        "status": status,
    }

# ---------------------------------------------------------------------
# Render a single close-up frame
# ---------------------------------------------------------------------

def _render_one_frame(i, output_directory):
    global _WORKER_SCENE

    if _WORKER_SCENE is None:
        _initialize_render_worker()

    d = _PARALLEL_DATA
    scene = _WORKER_SCENE
    ax = scene["ax"]

    x, y, z = float(d["x"][i]), float(d["y"][i]), float(d["z"][i])

    forward = np.array([d["hx"][i], d["hy"][i]], dtype=float)
    n = np.linalg.norm(forward)
    forward = forward / n if n > 1e-9 else np.array([1.0, 0.0])
    sideways = np.array([-forward[1], forward[0]])

    def p(longitudinal, lateral=0.0, height=0.0):
        return np.array([
            x + longitudinal * forward[0] + lateral * sideways[0],
            y + longitudinal * forward[1] + lateral * sideways[1],
            z + height,
        ])

    r = d["wheel_radius"]
    track = d["rear_track"]

    rear_left = p(d["rear_offset"], -track / 2, r)
    rear_right = p(d["rear_offset"], track / 2, r)
    front = p(d["front_offset"], 0.0, r)
    rear_mid = (rear_left + rear_right) / 2.0

    # Draw three distinct wheel circles.
    theta = np.linspace(0, 2 * np.pi, 48)
    for wheel, hub, centre in zip(
        scene["wheels"], scene["hubs"], [rear_left, rear_right, front]
    ):
        wheel.set_data_3d(
            centre[0] + r * np.cos(theta) * forward[0],
            centre[1] + r * np.cos(theta) * forward[1],
            centre[2] + r * np.sin(theta),
        )
        hub.set_data_3d([centre[0]], [centre[1]], [centre[2]])

    crank = p(-0.05 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE)
    seat = p(-0.42 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE)
    head_tube = p(0.40 * TRIKE_SCALE, 0, 0.95 * TRIKE_SCALE)
    handlebar = p(0.48 * TRIKE_SCALE, 0, 1.10 * TRIKE_SCALE)

    geometry = {
        "rear_axle": [rear_left, rear_right],
        "front_fork": [head_tube, front],
        "main_frame": [rear_mid, crank, head_tube],
        "seat_stay": [rear_mid, seat, crank],
        "handlebar": [head_tube, handlebar],
        "crank": [
            p(-0.18 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE),
            p(0.08 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE),
        ],
        "seat": [
            p(-0.55 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE),
            p(-0.30 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE),
        ],
    }

    hip = p(-0.38 * TRIKE_SCALE, 0, 1.30 * TRIKE_SCALE)
    shoulder = p(-0.10 * TRIKE_SCALE, 0, 1.80 * TRIKE_SCALE)
    head_centre = p(-0.08 * TRIKE_SCALE, 0, 2.02 * TRIKE_SCALE)
    hand = p(0.42 * TRIKE_SCALE, 0, 1.12 * TRIKE_SCALE)
    knee = p(-0.05 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE)
    foot = p(0.02 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE)

    geometry["torso"] = [hip, shoulder]
    geometry["arms"] = [shoulder, hand]
    geometry["legs"] = [hip, knee, foot]

    for name, points in geometry.items():
        arr = np.asarray(points)
        scene["segments"][name].set_data_3d(
            arr[:, 0], arr[:, 1], arr[:, 2]
        )

    scene["head"].set_data_3d(
        [head_centre[0]], [head_centre[1]], [head_centre[2]]
    )

    # Keep the recent trail visible.
    trail_start = max(0, i - VIDEO_FPS * 8)
    scene["trail"].set_data_3d(
        d["x"][trail_start:i + 1],
        d["y"][trail_start:i + 1],
        d["z"][trail_start:i + 1] + 1.5,
    )

    # Use the precomputed chronological camera angle. This is independent of
    # worker scheduling, so parallel rendering cannot make the camera stutter.
    azimuth = float(np.degrees(d["camera_azimuth"][i]))
    ax.view_init(elev=d["camera_elevation"], azim=azimuth)

    # Dynamic close-up horizontal limits. Clamp the view to the DEM bounds
    # so the camera does not drift into the axes background.
    cx = x + forward[0] * d["look_ahead"]
    cy = y + forward[1] * d["look_ahead"]

    half_w = d["camera_width"] / 2.0
    half_h = d["camera_depth"] / 2.0

    x0, x1 = cx - half_w, cx + half_w
    y0, y1 = cy - half_h, cy + half_h

    if x0 < d["terrain_xmin"]:
        x1 += d["terrain_xmin"] - x0
        x0 = d["terrain_xmin"]
    if x1 > d["terrain_xmax"]:
        x0 -= x1 - d["terrain_xmax"]
        x1 = d["terrain_xmax"]

    if y0 < d["terrain_ymin"]:
        y1 += d["terrain_ymin"] - y0
        y0 = d["terrain_ymin"]
    if y1 > d["terrain_ymax"]:
        y0 -= y1 - d["terrain_ymax"]
        y1 = d["terrain_ymax"]

    # Keep the camera window inside the DEM even near the edges.
    x0 = max(x0, d["terrain_xmin"])
    x1 = min(x1, d["terrain_xmax"])
    y0 = max(y0, d["terrain_ymin"])
    y1 = min(y1, d["terrain_ymax"])

    ax.set_xlim(x0, x1)
    ax.set_ylim(y0, y1)

    # Elevation window follows the vehicle. Clip to the terrain's overall
    # elevation range, while ensuring sufficient relief is visible.
    z0 = max(d["terrain_zmin"], z - d["local_z_below"])
    z1 = min(d["terrain_zmax"], z + d["local_z_above"])
    if z1 - z0 < 35.0:
        z0 = max(d["terrain_zmin"], z - 35.0)
        z1 = min(d["terrain_zmax"], z + 35.0)
    if z1 <= z0:
        z0, z1 = d["terrain_zmin"], d["terrain_zmax"]

    ax.set_zlim(z0, z1)

    # Explicit aspect ratio makes elevation changes much easier to see.
    ax.set_box_aspect((
        max(x1 - x0, 1.0),
        max(y1 - y0, 1.0),
        max((z1 - z0) * d["vertical_aspect_factor"], 1.0),
    ))

    t = float(d["times"][i])
    v = float(d["velocity"][i])
    s_value = float(d["distance"][i])

    scene["status"].set_text(
        f"Elapsed simulation time: {t:,.1f} s"
        f"     |     Velocity: {v:,.2f} m/s ({v * 3.6:,.1f} km/h)"
        f"     |     Distance: {s_value:,.0f} m"
    )

    frame_path = os.path.join(output_directory, f"frame_{i:06d}.jpg")
    scene["fig"].savefig(
        frame_path,
        dpi=d["video_dpi"],
        format="jpg",
        pil_kwargs={"quality": 92},
    )

    return int(i)

# ---------------------------------------------------------------------
# Parallel rendering with frame number and percentage
# ---------------------------------------------------------------------

ffmpeg_path = shutil.which("ffmpeg")
if ffmpeg_path is None:
    raise RuntimeError("FFmpeg was not found. Install it and ensure ffmpeg is on PATH.")

if "fork" not in mp.get_all_start_methods():
    raise RuntimeError("This cell requires Linux multiprocessing with the fork start method.")

with tempfile.TemporaryDirectory(prefix="castle_hill_frames_") as frame_dir:
    context = mp.get_context("fork")
    completed = 0
    last_line_length = 0

    with ProcessPoolExecutor(
        max_workers=MAX_WORKERS,
        mp_context=context,
        initializer=_initialize_render_worker,
    ) as executor:
        futures = {
            executor.submit(_render_one_frame, i, frame_dir): i
            for i in range(frame_count)
        }

        for future in as_completed(futures):
            finished_frame = future.result()
            completed += 1
            percentage = 100.0 * completed / frame_count

            message = (
                f"Completed frame {finished_frame + 1:,}/{frame_count:,}"
                f" | Overall progress: {percentage:6.2f}%"
                f" | Frames finished: {completed:,}/{frame_count:,}"
            )

            print(
                "\r" + message + " " * max(0, last_line_length - len(message)),
                end="",
                flush=True,
            )
            last_line_length = len(message)

    print("\nAll frames rendered. Encoding MP4...")

    subprocess.run(
        [
            ffmpeg_path,
            "-y",
            "-hide_banner",
            "-loglevel", "warning",
            "-framerate", str(VIDEO_FPS),
            "-i", os.path.join(frame_dir, "frame_%06d.jpg"),
            "-c:v", "libx264",
            "-preset", "medium",
            "-crf", str(VIDEO_CRF),
            "-pix_fmt", "yuv420p",
            "-threads", str(MAX_WORKERS),
            "-movflags", "+faststart",
            OUTPUT_FILE,
        ],
        check=True,
    )

print(f"\nFinished: {os.path.abspath(OUTPUT_FILE)}")
print(f"File size: {os.path.getsize(OUTPUT_FILE) / (1024 ** 2):,.1f} MiB")
```

```python
# COMPLETE SINGLE-CELL WORKFLOW: Mount Stuart descent.
# Fetch the Mount Stuart Road route and QldDem terrain, calculate a downhill
# speed profile, and render a smoothed-camera 3D animation.
#
# Requires: numpy, scipy, requests, matplotlib and ffmpeg.
# Run in a Linux Jupyter environment with an internet connection.

import json
import math
import os
import shutil
import subprocess
import tempfile
import multiprocessing as mp
import time
from concurrent.futures import ProcessPoolExecutor, as_completed

import requests
import numpy as np
from scipy.interpolate import PchipInterpolator
from scipy.integrate import solve_ivp
from scipy.signal import savgol_filter

import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib import cm


# =====================================================================
# 1. Fetch Mount Stuart Road from OpenStreetMap/OSRM
# =====================================================================

print("Fetching Mount Stuart Road route from OpenStreetMap/OSRM...")

def _geocode(address):
    response = requests.get(
        "https://nominatim.openstreetmap.org/search",
        params={
            "q": address,
            "format": "json",
            "countrycodes": "au",
            "limit": 10,
        },
        headers={"User-Agent": "Mount-Stuart-terrain-animation/1.0"},
        timeout=30,
    )
    response.raise_for_status()
    results = response.json()

    if not results:
        raise RuntimeError(f"No geocoding result for {address!r}")

    for item in results:
        name = item.get("display_name", "").lower()
        if "stuart drive" in name:
            return float(item["lon"]), float(item["lat"])

    raise RuntimeError(
        f"Could not identify Stuart Drive in geocoding results: {address!r}"
    )


# Approximate Mount Stuart summit lookout and Mount Stuart Road entrance.
# The summit is around 19.3432° S, 146.7810° E.
# The road entrance is around 19.3609° S, 146.8372° E.
summit_lon, summit_lat = 146.7810, -19.3432
base_lon, base_lat = 146.8372, -19.3609

_route_url = (
    "https://router.project-osrm.org/route/v1/driving/"
    f"{summit_lon},{summit_lat};{base_lon},{base_lat}"
)

_route_response = requests.get(
    _route_url,
    params={
        "overview": "full",
        "geometries": "geojson",
        "steps": "true",
    },
    timeout=60,
)
_route_response.raise_for_status()
_route_result = _route_response.json()

if _route_result.get("code") != "Ok":
    raise RuntimeError(f"OSRM routing failed: {_route_result}")

_route = _route_result["routes"][0]
_route_coords = np.asarray(_route["geometry"]["coordinates"], dtype=float)

_route_names = [
    step.get("name", "").strip()
    for leg in _route.get("legs", [])
    for step in leg.get("steps", [])
]

print("OSRM route roads:")
for _name in _route_names:
    if _name:
        print(f"  {_name}")

if not _route_names:
    raise RuntimeError("OSRM returned no named road segments.")

if not any(
    "stuart" in name.lower()
    for name in _route_names
):
    raise RuntimeError(
        "OSRM did not return a route containing a road with 'Stuart' "
        "in its name. Inspect the route before proceeding."
    )


# =====================================================================
# 2. Convert longitude/latitude to local metric coordinates
# =====================================================================

_center_lon = float(np.mean(_route_coords[:, 0]))
_center_lat = float(np.mean(_route_coords[:, 1]))

_m_per_deg_x = 111320.0 * math.cos(math.radians(_center_lat))
_m_per_deg_y = 110574.0


def _local_xy(lon, lat):
    return (
        (np.asarray(lon, dtype=float) - _center_lon) * _m_per_deg_x,
        (np.asarray(lat, dtype=float) - _center_lat) * _m_per_deg_y,
    )


# =====================================================================
# 3. Query the Queensland QldDem elevation service
# =====================================================================

# The identify endpoint matches the identify-style response expected below.
_identify_url = (
    "https://spatial-img.information.qld.gov.au/"
    "arcgis/rest/services/Elevation/QldDem/ImageServer/identify"
)


def _query_qlddem(lons, lats, chunk_size=100, max_retries=3):
    lons = np.asarray(lons, dtype=float).ravel()
    lats = np.asarray(lats, dtype=float).ravel()

    if len(lons) != len(lats):
        raise ValueError(
            f"Longitude and latitude arrays differ in length: "
            f"{len(lons)} != {len(lats)}"
        )

    if chunk_size < 1:
        raise ValueError("chunk_size must be at least 1")

    if max_retries < 0:
        raise ValueError("max_retries cannot be negative")

    values = np.full(len(lons), np.nan, dtype=float)
    total = len(lons)

    def _fetch_batch(begin, end):
        geometry = {
            "points": [
                [float(lon), float(lat)]
                for lon, lat in zip(lons[begin:end], lats[begin:end])
            ],
            "spatialReference": {"wkid": 4326},
        }

        params = {
            "geometry": json.dumps(geometry),
            "geometryType": "esriGeometryMultipoint",
            "returnGeometry": "false",
            "returnCatalogItems": "false",
            "f": "json",
        }

        last_error = None

        for attempt in range(max_retries + 1):
            try:
                response = requests.get(
                    _identify_url,
                    params=params,
                    timeout=60,
                )
                response.raise_for_status()
                payload = response.json()

                if "error" in payload:
                    raise RuntimeError(
                        f"QldDem identify error: {payload['error']}"
                    )

                results = payload.get("results")

                if not isinstance(results, list):
                    raise RuntimeError(
                        f"Unexpected QldDem identify response: {payload}"
                    )

                if len(results) != end - begin:
                    raise RuntimeError(
                        f"QldDem identify returned {len(results)} results "
                        f"for {end - begin} points"
                    )

                batch_values = np.full(end - begin, np.nan)

                for j, result in enumerate(results):
                    try:
                        batch_values[j] = float(result["value"])
                    except (KeyError, TypeError, ValueError):
                        pass

                if not np.all(np.isfinite(batch_values)):
                    raise RuntimeError(
                        "QldDem identify returned missing or non-finite "
                        "elevations."
                    )

                return batch_values

            except (
                requests.RequestException,
                ValueError,
                RuntimeError,
            ) as exc:
                last_error = exc

                if attempt < max_retries:
                    delay = min(2 ** attempt, 8)
                    print(
                        f"\nQldDem identify failed for samples "
                        f"{begin:,}–{end - 1:,} "
                        f"(attempt {attempt + 1}/{max_retries + 1}): "
                        f"{exc}. Retrying in {delay}s...",
                        flush=True,
                    )
                    time.sleep(delay)

        # Split a persistently failing batch to isolate problematic requests.
        if end - begin > 1:
            midpoint = begin + (end - begin) // 2
            left = _fetch_batch(begin, midpoint)
            right = _fetch_batch(midpoint, end)
            return np.concatenate((left, right))

        raise RuntimeError(
            f"QldDem identify failed for sample {begin}: "
            f"longitude={lons[begin]:.8f}, "
            f"latitude={lats[begin]:.8f}. Last error: {last_error}"
        ) from last_error

    for begin in range(0, total, chunk_size):
        end = min(begin + chunk_size, total)
        values[begin:end] = _fetch_batch(begin, end)

        print(
            f"\rQldDem samples: {end:,}/{total:,}",
            end="",
            flush=True,
        )

    print()

    if not np.all(np.isfinite(values)):
        raise RuntimeError(
            f"QldDem returned "
            f"{np.count_nonzero(~np.isfinite(values))} invalid elevations."
        )

    return values


# =====================================================================
# 4. Sample the route and construct its elevation profile
# =====================================================================

def _haversine(lon1, lat1, lon2, lat2):
    radius = 6371000.0

    p1 = math.radians(lat1)
    p2 = math.radians(lat2)
    dp = math.radians(lat2 - lat1)
    dl = math.radians(lon2 - lon1)

    a = (
        math.sin(dp / 2) ** 2
        + math.cos(p1) * math.cos(p2) * math.sin(dl / 2) ** 2
    )

    return 2 * radius * math.asin(math.sqrt(a))


_route_s = np.zeros(len(_route_coords), dtype=float)

for _i in range(1, len(_route_coords)):
    _route_s[_i] = (
        _route_s[_i - 1]
        + _haversine(
            *_route_coords[_i - 1],
            *_route_coords[_i],
        )
    )

# Mount Stuart Road is approximately 9 km long.
if not 6000.0 <= _route_s[-1] <= 12000.0:
    raise RuntimeError(
        f"Unexpected Mount Stuart route length: {_route_s[-1]:.1f} m. "
        "Check the OSRM route before querying the terrain."
    )

_profile_s = np.arange(0.0, _route_s[-1], 10.0)

if len(_profile_s) == 0 or _profile_s[-1] < _route_s[-1]:
    _profile_s = np.append(_profile_s, _route_s[-1])

_route_lon = np.interp(
    _profile_s,
    _route_s,
    _route_coords[:, 0],
)
_route_lat = np.interp(
    _profile_s,
    _route_s,
    _route_coords[:, 1],
)

print(f"Route length: {_route_s[-1] / 1000:.3f} km")
print("Querying QldDem for the route elevation profile...")

# Small chunks preserve the conservative request behaviour of the original.
_profile_z = _query_qlddem(
    _route_lon,
    _route_lat,
    chunk_size=3,
)

if len(_profile_z) >= 11:
    _profile_z = savgol_filter(_profile_z, 11, 3)

# OSRM was queried summit -> base, so the route is already oriented for descent.
distance = _profile_s.copy()
h = _profile_z.copy()

road_lon = _route_lon.copy()
road_lat = _route_lat.copy()

road_x, road_y = _local_xy(road_lon, road_lat)
road_z = h.copy()


# =====================================================================
# 5. Build the surrounding QldDem terrain mesh
# =====================================================================

_grid_spacing = 65.0
_margin = 1800.0

_xmin = float(np.min(road_x) - _margin)
_xmax = float(np.max(road_x) + _margin)
_ymin = float(np.min(road_y) - _margin)
_ymax = float(np.max(road_y) + _margin)

_grid_x = np.arange(
    _xmin,
    _xmax + _grid_spacing,
    _grid_spacing,
)
_grid_y = np.arange(
    _ymin,
    _ymax + _grid_spacing,
    _grid_spacing,
)

terrain_x, terrain_y = np.meshgrid(_grid_x, _grid_y)

_grid_lon = terrain_x.ravel() / _m_per_deg_x + _center_lon
_grid_lat = terrain_y.ravel() / _m_per_deg_y + _center_lat

print(
    f"Querying QldDem for terrain mesh "
    f"({terrain_x.shape[1]} x {terrain_x.shape[0]} points)..."
)

terrain_z = _query_qlddem(
    _grid_lon,
    _grid_lat,
    chunk_size=3,
).reshape(terrain_x.shape)


# =====================================================================
# 6. Calculate the coasting speed profile
# =====================================================================

mass = 64.0 + 20.0 + 5.0
gravity = 9.81
air_density = 1.225
rolling_resistance = 0.006
drag_area = 0.50

slope = np.gradient(h, distance)

# Positive sin_theta means downhill acceleration.
sin_theta = np.clip(-slope, -1.0, 1.0)
cos_theta = np.sqrt(np.maximum(0.0, 1.0 - sin_theta**2))

sin_fn = PchipInterpolator(distance, sin_theta, extrapolate=True)
cos_fn = PchipInterpolator(distance, cos_theta, extrapolate=True)


def _speed_ode(s_pos, w):
    st = float(sin_fn(s_pos))
    ct = float(cos_fn(s_pos))
    w0 = max(float(w[0]), 0.0)

    acceleration = (
        gravity * st
        - rolling_resistance * gravity * ct
        - air_density * drag_area / (2 * mass) * w0
    )

    if w0 <= 0.0 and acceleration < 0.0:
        return [0.0]

    return [2.0 * acceleration]


_solution = solve_ivp(
    _speed_ode,
    (float(distance[0]), float(distance[-1])),
    [0.0],
    t_eval=distance,
    rtol=1e-8,
    atol=1e-10,
    max_step=2.0,
)

if not _solution.success:
    raise RuntimeError(f"Speed integration failed: {_solution.message}")

velocity = np.sqrt(np.maximum(_solution.y[0], 0.0))

print(
    f"Prepared Mount Stuart descent: {distance[-1] / 1000:.3f} km; "
    f"elevation {h[0]:.1f} m -> {h[-1]:.1f} m; "
    f"peak speed {np.max(velocity) * 3.6:.1f} km/h"
)
print("Starting the Mount Stuart animation render...")


# =====================================================================
# 7. Animation configuration
# =====================================================================

OUTPUT_FILE = "mount_stuart_terrain_descent.mp4"

MAX_WORKERS = min(8, os.cpu_count() or 1)
VIDEO_FPS = 10
VIDEO_DPI = 100
VIDEO_CRF = 18
TERRAIN_STRIDE = 3

CAMERA_WIDTH = 700.0
CAMERA_DEPTH = 550.0
CAMERA_LOOK_AHEAD = 100.0
CAMERA_ELEVATION = 38.0
CAMERA_TURN_SMOOTHING = 0.20

LOCAL_Z_BELOW = 65.0
LOCAL_Z_ABOVE = 95.0
VERTICAL_ASPECT_FACTOR = 2.5

TRIKE_SCALE = 4.0
WHEEL_RADIUS = 0.34 * TRIKE_SCALE
REAR_TRACK = 0.85 * TRIKE_SCALE
REAR_OFFSET = -0.85 * TRIKE_SCALE
FRONT_OFFSET = 0.90 * TRIKE_SCALE


# =====================================================================
# 8. Validate and prepare the animation data
# =====================================================================

required = (
    "terrain_x", "terrain_y", "terrain_z",
    "road_x", "road_y", "road_z",
    "distance", "velocity",
)

missing = [name for name in required if name not in globals()]

if missing:
    raise NameError(
        "Missing notebook variables: " + ", ".join(missing)
    )

tx = np.asarray(terrain_x, dtype=float)
ty = np.asarray(terrain_y, dtype=float)
tz = np.asarray(terrain_z, dtype=float)

rx = np.asarray(road_x, dtype=float).ravel()
ry = np.asarray(road_y, dtype=float).ravel()
rz = np.asarray(road_z, dtype=float).ravel()

dist = np.asarray(distance, dtype=float).ravel()
vel = np.asarray(velocity, dtype=float).ravel()

if tx.ndim != 2 or tx.shape != ty.shape or tx.shape != tz.shape:
    raise ValueError(
        "terrain_x, terrain_y and terrain_z must have matching 2D shapes."
    )

if not (len(rx) == len(ry) == len(rz)):
    raise ValueError("Road arrays must have equal lengths.")

if len(rx) < 2 or len(dist) < 2 or len(dist) != len(vel):
    raise ValueError("The route and simulation arrays are incomplete.")

valid = np.isfinite(dist) & np.isfinite(vel)
dist, vel = dist[valid], vel[valid]

order = np.argsort(dist)
dist, vel = dist[order], vel[order]

dist, unique_idx = np.unique(dist, return_index=True)
vel = vel[unique_idx]

if len(dist) < 2 or dist[-1] <= dist[0]:
    raise ValueError("Simulation distance must be strictly increasing.")

if not (
    np.all(np.isfinite(rx))
    and np.all(np.isfinite(ry))
    and np.all(np.isfinite(rz))
):
    raise ValueError("The road arrays contain non-finite coordinates.")

# The upstream route is already summit -> Stuart Drive. Only reverse it
# if its elevations clearly show the opposite orientation.
reverse_route = rz[0] < rz[-1]

if reverse_route:
    rx = rx[::-1].copy()
    ry = ry[::-1].copy()
    rz = rz[::-1].copy()
    print("Reversed road coordinates to match the descent direction.")

route_ds = np.hypot(np.diff(rx), np.diff(ry))
route_s = np.concatenate(([0.0], np.cumsum(route_ds)))

route_s, route_unique_idx = np.unique(
    route_s,
    return_index=True,
)

rx = rx[route_unique_idx]
ry = ry[route_unique_idx]
rz = rz[route_unique_idx]

route_length = float(route_s[-1])
simulation_length = float(dist[-1] - dist[0])

if route_length <= 0 or simulation_length <= 0:
    raise ValueError(
        "Could not determine a positive route or simulation length."
    )

route_progress = (
    (dist - dist[0]) / simulation_length * route_length
)

path_x = np.interp(route_progress, route_s, rx)
path_y = np.interp(route_progress, route_s, ry)
path_z = np.interp(route_progress, route_s, rz)

safe_velocity = np.maximum(vel, 0.1)

elapsed = np.zeros_like(dist)
elapsed[1:] = np.cumsum(
    np.diff(dist) * 0.5 * (
        1.0 / safe_velocity[:-1]
        + 1.0 / safe_velocity[1:]
    )
)

duration = float(elapsed[-1])

if not np.isfinite(duration) or duration <= 0:
    raise ValueError("Could not calculate a valid simulation duration.")

frame_count = max(
    2,
    int(np.ceil(duration * VIDEO_FPS)) + 1,
)

frame_times = np.linspace(0.0, duration, frame_count)

frame_dist = np.interp(frame_times, elapsed, dist)
frame_vel = np.interp(frame_times, elapsed, vel)
frame_x = np.interp(frame_dist, dist, path_x)
frame_y = np.interp(frame_dist, dist, path_y)
frame_z = np.interp(frame_dist, dist, path_z)

frame_x[0], frame_y[0], frame_z[0] = path_x[0], path_y[0], path_z[0]
frame_x[-1], frame_y[-1], frame_z[-1] = path_x[-1], path_y[-1], path_z[-1]

heading_x = np.gradient(frame_x)
heading_y = np.gradient(frame_y)

heading_norm = np.hypot(heading_x, heading_y)
heading_norm[heading_norm < 1e-9] = 1.0

heading_x /= heading_norm
heading_y /= heading_norm

raw_azimuth = np.unwrap(
    np.arctan2(heading_y, heading_x) + np.pi
)

heading_window = max(3, int(round(VIDEO_FPS * 0.8)))

if heading_window % 2 == 0:
    heading_window += 1

heading_kernel = np.ones(heading_window, dtype=float) / heading_window
pad = heading_window // 2

padded_azimuth = np.pad(
    raw_azimuth,
    (pad, pad),
    mode="edge",
)

smoothed_azimuth = np.convolve(
    padded_azimuth,
    heading_kernel,
    mode="valid",
)

camera_azimuth = np.empty_like(smoothed_azimuth)
camera_azimuth[0] = smoothed_azimuth[0]

for frame_i in range(1, len(camera_azimuth)):
    camera_azimuth[frame_i] = (
        camera_azimuth[frame_i - 1]
        + CAMERA_TURN_SMOOTHING
        * (
            smoothed_azimuth[frame_i]
            - camera_azimuth[frame_i - 1]
        )
    )

finite_xy = np.isfinite(tx) & np.isfinite(ty)

terrain_xmin = float(np.min(tx[finite_xy]))
terrain_xmax = float(np.max(tx[finite_xy]))
terrain_ymin = float(np.min(ty[finite_xy]))
terrain_ymax = float(np.max(ty[finite_xy]))

finite_z = tz[np.isfinite(tz)]

if len(finite_z) == 0:
    raise ValueError("The terrain has no finite elevation values.")

terrain_zmin = float(np.percentile(finite_z, 1))
terrain_zmax = float(np.percentile(finite_z, 99))

print(f"Terrain array: {tz.shape}")
print(f"Route length: {route_length:,.1f} m")
print(f"Simulation duration: {duration:,.1f} s")
print(f"Frames: {frame_count:,}")
print(f"Start: ({frame_x[0]:.1f}, {frame_y[0]:.1f}, {frame_z[0]:.1f})")
print(f"Finish: ({frame_x[-1]:.1f}, {frame_y[-1]:.1f}, {frame_z[-1]:.1f})")
print(f"Rendering with up to {MAX_WORKERS} CPU cores.")


_PARALLEL_DATA = {
    "terrain_x": tx[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "terrain_y": ty[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "terrain_z": tz[::TERRAIN_STRIDE, ::TERRAIN_STRIDE],
    "road_x": rx,
    "road_y": ry,
    "road_z": rz,
    "times": frame_times,
    "distance": frame_dist,
    "velocity": frame_vel,
    "x": frame_x,
    "y": frame_y,
    "z": frame_z,
    "hx": heading_x,
    "hy": heading_y,
    "camera_azimuth": camera_azimuth,
    "frame_count": frame_count,
    "terrain_xmin": terrain_xmin,
    "terrain_xmax": terrain_xmax,
    "terrain_ymin": terrain_ymin,
    "terrain_ymax": terrain_ymax,
    "terrain_zmin": terrain_zmin,
    "terrain_zmax": terrain_zmax,
    "camera_width": CAMERA_WIDTH,
    "camera_depth": CAMERA_DEPTH,
    "look_ahead": CAMERA_LOOK_AHEAD,
    "camera_elevation": CAMERA_ELEVATION,
    "local_z_below": LOCAL_Z_BELOW,
    "local_z_above": LOCAL_Z_ABOVE,
    "vertical_aspect_factor": VERTICAL_ASPECT_FACTOR,
    "wheel_radius": WHEEL_RADIUS,
    "rear_track": REAR_TRACK,
    "rear_offset": REAR_OFFSET,
    "front_offset": FRONT_OFFSET,
    "video_dpi": VIDEO_DPI,
}

_WORKER_SCENE = None


# =====================================================================
# 9. Initialize a Matplotlib scene in each worker
# =====================================================================

def _initialize_render_worker():
    global _WORKER_SCENE

    d = _PARALLEL_DATA

    fig = plt.figure(figsize=(13.0, 9.0), dpi=d["video_dpi"])
    ax = fig.add_subplot(111, projection="3d")

    zz = d["terrain_z"]

    zmin = float(np.nanpercentile(zz, 2))
    zmax = float(np.nanpercentile(zz, 98))

    if zmax <= zmin:
        zmax = zmin + 1.0

    terrain_colours = cm.terrain(
        mpl.colors.Normalize(vmin=zmin, vmax=zmax)(zz)
    )
    terrain_colours[..., 3] = np.where(
        np.isfinite(zz),
        1.0,
        0.0,
    )

    ax.plot_surface(
        d["terrain_x"],
        d["terrain_y"],
        zz,
        facecolors=terrain_colours,
        linewidth=0,
        antialiased=False,
        shade=True,
    )

    ax.plot(
        d["road_x"],
        d["road_y"],
        d["road_z"] + 0.8,
        color="black",
        linewidth=3.2,
        zorder=15,
    )

    ax.plot(
        d["road_x"],
        d["road_y"],
        d["road_z"] + 1.6,
        color="white",
        linewidth=1.1,
        zorder=16,
    )

    trail, = ax.plot(
        [],
        [],
        [],
        color="orangered",
        linewidth=3.0,
        zorder=25,
    )

    names = [
        "rear_axle", "front_fork", "main_frame", "seat_stay",
        "handlebar", "crank", "seat", "torso", "arms", "legs",
    ]

    segments = {}

    for name in names:
        colour = "#ff8c00"
        width = 4.5

        if name in ("torso", "arms", "legs"):
            colour = "#b21f24"
            width = 5.5

        line, = ax.plot(
            [],
            [],
            [],
            color=colour,
            linewidth=width,
            solid_capstyle="round",
            zorder=40,
        )

        segments[name] = line

    wheels, hubs = [], []

    for _ in range(3):
        wheel, = ax.plot(
            [],
            [],
            [],
            color="#111111",
            linewidth=3.5,
            zorder=42,
        )
        wheels.append(wheel)

        hub, = ax.plot(
            [],
            [],
            [],
            marker="o",
            markersize=5,
            linestyle="None",
            color="white",
            markeredgecolor="black",
            markeredgewidth=1.2,
            zorder=44,
        )
        hubs.append(hub)

    head, = ax.plot(
        [],
        [],
        [],
        marker="o",
        markersize=11,
        linestyle="None",
        markerfacecolor="#ffd28a",
        markeredgecolor="black",
        markeredgewidth=1.5,
        zorder=48,
    )

    fig.text(
        0.5,
        0.975,
        "Mount Stuart Road descent",
        ha="center",
        va="top",
        fontsize=16,
        fontweight="bold",
    )

    status = fig.text(
        0.5,
        0.935,
        "",
        ha="center",
        va="top",
        fontsize=11,
    )

    fig.subplots_adjust(
        left=0.04,
        right=0.88,
        bottom=0.12,
        top=0.89,
    )

    ax.set_xlabel("West–east position (m)", labelpad=10)
    ax.set_ylabel("South–north position (m)", labelpad=10)
    ax.set_zlabel("Elevation (m)", labelpad=8)
    ax.set_facecolor("#eeeeee")
    ax.grid(True)
    ax.set_proj_type("persp", focal_length=0.9)

    _WORKER_SCENE = {
        "fig": fig,
        "ax": ax,
        "trail": trail,
        "segments": segments,
        "wheels": wheels,
        "hubs": hubs,
        "head": head,
        "status": status,
    }


# =====================================================================
# 10. Render an individual animation frame
# =====================================================================

def _render_one_frame(i, output_directory):
    global _WORKER_SCENE

    if _WORKER_SCENE is None:
        _initialize_render_worker()

    d = _PARALLEL_DATA
    scene = _WORKER_SCENE
    ax = scene["ax"]

    x = float(d["x"][i])
    y = float(d["y"][i])
    z = float(d["z"][i])

    forward = np.array(
        [d["hx"][i], d["hy"][i]],
        dtype=float,
    )

    n = np.linalg.norm(forward)

    if n > 1e-9:
        forward /= n
    else:
        forward = np.array([1.0, 0.0])

    sideways = np.array([-forward[1], forward[0]])

    def p(longitudinal, lateral=0.0, height=0.0):
        return np.array([
            x + longitudinal * forward[0] + lateral * sideways[0],
            y + longitudinal * forward[1] + lateral * sideways[1],
            z + height,
        ])

    r = d["wheel_radius"]
    track = d["rear_track"]

    rear_left = p(d["rear_offset"], -track / 2, r)
    rear_right = p(d["rear_offset"], track / 2, r)
    front = p(d["front_offset"], 0.0, r)
    rear_mid = (rear_left + rear_right) / 2.0

    theta = np.linspace(0, 2 * np.pi, 48)

    for wheel, hub, centre in zip(
        scene["wheels"],
        scene["hubs"],
        [rear_left, rear_right, front],
    ):
        wheel.set_data_3d(
            centre[0] + r * np.cos(theta) * forward[0],
            centre[1] + r * np.cos(theta) * forward[1],
            centre[2] + r * np.sin(theta),
        )

        hub.set_data_3d(
            [centre[0]],
            [centre[1]],
            [centre[2]],
        )

    crank = p(
        -0.05 * TRIKE_SCALE,
        0,
        0.65 * TRIKE_SCALE,
    )

    seat = p(
        -0.42 * TRIKE_SCALE,
        0,
        1.02 * TRIKE_SCALE,
    )

    head_tube = p(
        0.40 * TRIKE_SCALE,
        0,
        0.95 * TRIKE_SCALE,
    )

    handlebar = p(
        0.48 * TRIKE_SCALE,
        0,
        1.10 * TRIKE_SCALE,
    )

    geometry = {
        "rear_axle": [rear_left, rear_right],
        "front_fork": [head_tube, front],
        "main_frame": [rear_mid, crank, head_tube],
        "seat_stay": [rear_mid, seat, crank],
        "handlebar": [head_tube, handlebar],
        "crank": [
            p(-0.18 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE),
            p(0.08 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE),
        ],
        "seat": [
            p(-0.55 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE),
            p(-0.30 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE),
        ],
    }

    hip = p(-0.38 * TRIKE_SCALE, 0, 1.30 * TRIKE_SCALE)
    shoulder = p(-0.10 * TRIKE_SCALE, 0, 1.80 * TRIKE_SCALE)
    head_centre = p(-0.08 * TRIKE_SCALE, 0, 2.02 * TRIKE_SCALE)
    hand = p(0.42 * TRIKE_SCALE, 0, 1.12 * TRIKE_SCALE)
    knee = p(-0.05 * TRIKE_SCALE, 0, 1.02 * TRIKE_SCALE)
    foot = p(0.02 * TRIKE_SCALE, 0, 0.65 * TRIKE_SCALE)

    geometry["torso"] = [hip, shoulder]
    geometry["arms"] = [shoulder, hand]
    geometry["legs"] = [hip, knee, foot]

    for name, points in geometry.items():
        arr = np.asarray(points)

        scene["segments"][name].set_data_3d(
            arr[:, 0],
            arr[:, 1],
            arr[:, 2],
        )

    scene["head"].set_data_3d(
        [head_centre[0]],
        [head_centre[1]],
        [head_centre[2]],
    )

    trail_start = max(0, i - VIDEO_FPS * 8)

    scene["trail"].set_data_3d(
        d["x"][trail_start:i + 1],
        d["y"][trail_start:i + 1],
        d["z"][trail_start:i + 1] + 1.5,
    )

    azimuth = float(np.degrees(d["camera_azimuth"][i]))

    ax.view_init(
        elev=d["camera_elevation"],
        azim=azimuth,
    )

    cx = x + forward[0] * d["look_ahead"]
    cy = y + forward[1] * d["look_ahead"]

    half_w = d["camera_width"] / 2.0
    half_h = d["camera_depth"] / 2.0

    x0, x1 = cx - half_w, cx + half_w
    y0, y1 = cy - half_h, cy + half_h

    if x0 < d["terrain_xmin"]:
        x1 += d["terrain_xmin"] - x0
        x0 = d["terrain_xmin"]

    if x1 > d["terrain_xmax"]:
        x0 -= x1 - d["terrain_xmax"]
        x1 = d["terrain_xmax"]

    if y0 < d["terrain_ymin"]:
        y1 += d["terrain_ymin"] - y0
        y0 = d["terrain_ymin"]

    if y1 > d["terrain_ymax"]:
        y0 -= y1 - d["terrain_ymax"]
        y1 = d["terrain_ymax"]

    x0 = max(x0, d["terrain_xmin"])
    x1 = min(x1, d["terrain_xmax"])
    y0 = max(y0, d["terrain_ymin"])
    y1 = min(y1, d["terrain_ymax"])

    ax.set_xlim(x0, x1)
    ax.set_ylim(y0, y1)

    z0 = max(d["terrain_zmin"], z - d["local_z_below"])
    z1 = min(d["terrain_zmax"], z + d["local_z_above"])

    if z1 - z0 < 35.0:
        z0 = max(d["terrain_zmin"], z - 35.0)
        z1 = min(d["terrain_zmax"], z + 35.0)

    if z1 <= z0:
        z0 = d["terrain_zmin"]
        z1 = d["terrain_zmax"]

    ax.set_zlim(z0, z1)

    ax.set_box_aspect((
        max(x1 - x0, 1.0),
        max(y1 - y0, 1.0),
        max(
            (z1 - z0) * d["vertical_aspect_factor"],
            1.0,
        ),
    ))

    t = float(d["times"][i])
    v = float(d["velocity"][i])
    s_value = float(d["distance"][i])

    scene["status"].set_text(
        f"Elapsed simulation time: {t:,.1f} s"
        f"     |     Velocity: {v:,.2f} m/s ({v * 3.6:,.1f} km/h)"
        f"     |     Distance: {s_value:,.0f} m"
    )

    frame_path = os.path.join(
        output_directory,
        f"frame_{i:06d}.jpg",
    )

    scene["fig"].savefig(
        frame_path,
        dpi=d["video_dpi"],
        format="jpg",
        pil_kwargs={"quality": 92},
    )

    return int(i)


# =====================================================================
# 11. Render frames in parallel and encode the MP4
# =====================================================================

ffmpeg_path = shutil.which("ffmpeg")

if ffmpeg_path is None:
    raise RuntimeError(
        "FFmpeg was not found. Install it and ensure ffmpeg is on PATH."
    )

if "fork" not in mp.get_all_start_methods():
    raise RuntimeError(
        "This cell requires Linux multiprocessing with the fork start method."
    )

with tempfile.TemporaryDirectory(
    prefix="mount_stuart_frames_"
) as frame_dir:

    context = mp.get_context("fork")
    completed = 0
    last_line_length = 0

    with ProcessPoolExecutor(
        max_workers=MAX_WORKERS,
        mp_context=context,
        initializer=_initialize_render_worker,
    ) as executor:

        futures = {
            executor.submit(
                _render_one_frame,
                i,
                frame_dir,
            ): i
            for i in range(frame_count)
        }

        for future in as_completed(futures):
            finished_frame = future.result()
            completed += 1
            percentage = 100.0 * completed / frame_count

            message = (
                f"Completed frame {finished_frame + 1:,}/{frame_count:,}"
                f" | Overall progress: {percentage:6.2f}%"
                f" | Frames finished: {completed:,}/{frame_count:,}"
            )

            print(
                "\r" + message
                + " " * max(0, last_line_length - len(message)),
                end="",
                flush=True,
            )

            last_line_length = len(message)

    print("\nAll frames rendered. Encoding MP4...")

    subprocess.run(
        [
            ffmpeg_path,
            "-y",
            "-hide_banner",
            "-loglevel", "warning",
            "-framerate", str(VIDEO_FPS),
            "-i", os.path.join(frame_dir, "frame_%06d.jpg"),
            "-c:v", "libx264",
            "-preset", "medium",
            "-crf", str(VIDEO_CRF),
            "-pix_fmt", "yuv420p",
            "-threads", str(MAX_WORKERS),
            "-movflags", "+faststart",
            OUTPUT_FILE,
        ],
        check=True,
    )

print(f"\nFinished: {os.path.abspath(OUTPUT_FILE)}")
print(f"File size: {os.path.getsize(OUTPUT_FILE) / (1024 ** 2):,.1f} MiB")
```
